# Ficheiros

Os ficheiros permitem **guardar objetos/informação de forma permanente** e de fácil acesso.

O `Python` tem um conjunto rico de instruções para manipular ficheiros.

## Abrir / criar um ficheiro



### Método 1

Um objeto do tipo ficheiro é criado com a instrução `open`. Esta instrução abre o ficheiro com o nome indicado (na pasta em que estamos a trabalhar), caso contrário, cria um ficheiro com este nome . A partir daí, o ficheiro pode ser chamado por uma variável no programa. Um ficheiro aberto desta forma deve ser fechado depois de utilizado, para garantir que o ficheiro não fica corrompido:

````python
f = open('ficheiro.txt', mode='w')     # abre/cria o ficheiro no modo de escrita e coloca-o na variável f

f.close()   # fecha o ficheiro quando não queremos fazer mais nenhuma operação
````

### Método 2

Também podemos abrir um ficheiro usando outra sintaxe.

Com este método, o programa fecha o ficheiro que abriu no final do conjunto de instruções (bloco identado), logo não precisamos de o fechar.

````python
with open('ficheiro2.txt', mode='w') as f:       # não é necessário fechar o ficheiro com f.close()
    print('Criei / abri o ficheiro')        # todas as instruções identadas são realizadas com o ficheiro aberto
````

## Modos de abrir um ficheiro

Podemos abrir um ficheiro em modos diferentes, dependendo das operações que queremos aplicar:

- Modo de leitura `'r'`: para quando queremos ler as informações de um ficheiro.
- Modo de escrita `'w'`: para quando queremos escrever um ficheiro.
- Modo de apêndice `'a'`: para quando queremos adicionar informação ao fim de um ficheiro.
- Modo combinado de leitura e escrita `'r+'`: para quando  queremos ler e escrever no ficheiro ao mesmo tempo.

## Diretoria de um ficheiro

Com este tipo de instruções estamos a assumir que o ficheiro a que queremos aceder está na mesma diretoria onde está instalado o programa `Python`. Se não for o caso, temos que indicar o caminho para o ficheiro:

````python
with open('sample_data/mnist_test.csv','r') as f:
	print('Abri o ficheiro')
````

## Modificar um Ficheiro



### Modo de escrita, `'w'`

Para escrevermos informação num ficheiro podemos usar diversos métodos:


#### `write()`

Escreve uma única linha (`string`) no ficheiro:

````python
f = open('ficheiro.txt', mode='w')

f.write('Linha 1')
f.write('Linha 2')

f.close()
````

#### `writelines()`

Escreve uma lista de strings no ficheiro (sem parágrafo entre elas).

````python
f = open('ficheiro.txt', mode='w')

f.writelines(['Linha 1', 'Linha 2', 'Linha 3'])

f.close()
````

#### Mudança de linha

Se quisermos acrescentar um parágrafo entre cada linha, temos que colocar o caracter `\n` para indicar um parágrafo:

````python
f = open('ficheiro.txt', mode='w')

f.writelines(['Linha 1 \n', 'Linha 2 \n', 'Linha 3 \n'])

f.close()
````

### Modo de apêndice, `'a'`

Já vimos que quando abrimos o mesmo ficheiro no modo de escrita, ele escreve por cima do ficheiro original. Nos casos em que queremos acrescentar informação sem apagar o conteúdo original do ficheiro, temos que o abrir no modo de apêndice, `'a'`:

````python
with open('ficheiro.txt', mode = 'a') as f:
    f.write('\nLinha 4 \n')
    f.write('Linha 5')
````

### Modo de leitura, `'r'`

Para ler a informação de um ficheiro, podemos fazê-lo de várias formas:



#### Com um iterador `for`

Um ficheiro é um objeto iterável onde o iterador devolve, em sequência, cada linha do ficheiro:

````python
f = open('ficheiro.txt', 'r')

for line in f:
	print(line)

f.close()
````

#### Método `read()`

Lê `n` caracteres do ficheiro devolvendo-os numa `string`.

Se `n` for omitido, todo o ficheiro é lido.

````python
f = open('ficheiro.txt', 'r')

s = f.read()    # lê n caracteres do ficheiro devolvendo-os numa string.

print(s)

f.close()
````

#### Método `readlines()`

Lê todas as linhas do ficheiro e devolve-as numa lista de `strings` (uma `string` por linha):

````python
texto = open('ficheiro.txt', mode = 'r')

linhas = texto.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

texto.close()

print(linhas)
````

#### Saltar cabeçalhos

Em muitos casos, os ficheiros têm uma (ou várias) linhas de cabeçalho que não queremos ler. Para saltar linhas específicas na leitura de um ficheiro podemos fazer:

````python
f = open('ficheiro.txt', mode='r') # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

for line in linhas[1:]:     # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)
    print(line)

f.close()
````

# Operações importantes com `strings`

Cada linha de um ficheiro é uma string das quais vamos querer retirar informação para o nosso programa. Existem alguns métodos que nos ajudam muito a retirar esta informação de forma mais simples:


## Operação `strip()`

Este método retira certos caracteres do início ou do fim de uma `string`. É muito útil para retirar os `\n` dos ficheiros.

````python
linha_1 = 'x, y\n'
linha_2 = '3, 4\n'

print(linha_2)
print(linha_2.strip('\n'))
print(linha_2.strip('\n'))
````

## Operação `rstrip()` e `lstrip()`

As operações `rstrip()` e `lstrip()` são usadas para retirar caracteres de uma `string` à direita e à esquerda, respectivamente.

````python
s = '.O rato roeu a garrafa do Rei da Rússia.'

print(s.rstrip('.')) # Ex.: remover pontos finais
print(s.lstrip('.')) # Ex.: remover travessões
print(s.strip('.'))
````

## Operação `split()`

Transforma a string numa lista, com os elementos separados pelo caracter indicado. É muito útil para separar as colunas de dados de um ficheiro.

````python
linha_1 = 'x, y'
linha_2 = '3, 4'

linha = linha_2.split(',')

print(linha)
print(type(linha))
````

## Operação `replace()`

Para substituir um certo caracter.

````python
s = 'Eu vou comer laranjas e bananas.'

print(s.replace('a', 'e'))
````

## Alinhamento de `strings`

As funções `rjust(n, c)`, `ljust(n, c)`, e `center(n, c)` são úteis para escrever ficheiros bonitos.

Eles ajustam uma string à direita esquerda ou centro, dentro de um espaço de `n` caracteres, usando o caracter `c` como espaçamento. O caracter `c` é opcional e por _default_ é um ` `.

````python
s = '123'

#print(s.rjust(10, '*'))  # ajusta à direita de uma string de colocando * nos espaços em branco

#print(s.ljust(10, '-'))   # ajusta à esquerda colocando - nos espaços em branco
print('|', end='')
print(s.center(10), end='')
print('|', end='')
````

## Método `join()`

O método `join()` permite-nos criar uma `string` a partir de uma lista de `strings`.

````python
L = ['Python ', 'is ', 'a ', 'programming ', 'language.']
s = ''

string = s.join(L)

print(string)
print(type(string))
````

Escreva um programa que simule 10000 lançamentos de dois dados e, para cada lançamento, escreva uma linha com os valores dos dois dados (`1` a `6`) e a soma dos pontos dos dados (`2` a `12`) num ficheiro chamado `dados.txt`. Separe os valores escritos no ficheiro por uma vírgula.

````python
#@title Solução

from random import randint

f = open('dados.txt', mode='w')     # cria o ficheiro

f.write('Dado_1, Dado_2, Soma \n')  # escreve o cabeçalho

for i in range(10000):  # Para cada lançamento:

    # gerar os números aleatórios e calcular a soma
    dado_1 = randint(1,6)
    dado_2 = randint(1,6)
    soma = dado_1 + dado_2

    # construir a linha que queremos escrever em formato de string
    linha = str(dado_1)+', '+str(dado_2)+', '+str(soma)+'\n'
    f.write(linha)  # escrever a linha no ficheiro

f.close()   # fechar o ficheiro
````

### b) Ler o Ficheiro

_(Ficha 6, Problema 2)_

Escreva um programa que leia os dados do ficheiro `dados.txt` criado pelo programa do problema anterior e faça uma análise estatística determinando quantas vezes cada uma das somas possíveis dos pontos de dois dados ocorreu.

Escreva esses resultados num ficheiro designado `analise.txt`.

````python
#@title Solução

f = open('dados.txt', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

D = {} # dicionário vazio que vai contar quantas vezes ocorreu cada soma

for line in linhas[1:]:         # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)
    linha = line.split(',')     # dividimos em colunas
    soma = int(linha[-1].strip('\n'))   # convertemos a soma num número inteiro

    if soma not in D:   # se essa soma ainda não estiver no dicionário, a contagem é 1
        D[soma] = 1
    else:               # se a soma já estiver no dicionário, acrescentar uma unidade
        D[soma] = D[soma]+1

f.close()

output = open('analise.txt', mode = 'w')    # criar o ficheiro de output no modo de escrita

output.write('Soma, Contagens \n')      # escrever o cabeçalho

for soma in sorted(D.keys()):  # para cada chave do dicionário (ordenada por ordem crescente)
    linha = str(soma)+', '+str(D[soma])+'\n' # compor a linha a escrever
    output.write(linha)                     # escrever a linha no ficheiro

output.close()
````

## 2. Bagagem de avião

_(Ficha 6, Problema 3)_

O ficheiro `bagagem.txt` contem o peso em kg de 500 malas embarcadas num avião. Comece por colocar este ficheiro na mesma diretoria onde tem o programa e abrir este ficheiro para perceber como está organizada cada linha do mesmo.

Escreva um programa que leia o ficheiro e calcule o peso médio das malas, o desvio padrão do peso e o número de malas que excede o limite de peso de 22 kg.

````python
#@title Solução

f = open('bagagem.txt', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

Limite_peso = [] # lista vazia com as malas acima do peso

for line in linhas[1:]:         # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)
    peso = float(line.strip('\n'))      # retira o peso

    if peso > 22:                       # se o peso estiver acima do limite, coloca-o na lista
        Limite_peso.append(peso)

f.close()

print('Há', len(Limite_peso), 'malas acima do limite de peso (22kg).')


# Cálculo da média
media  = 0
for peso in Limite_peso:
    media = media + peso

media = media / len(Limite_peso)

print ('Estas malas têm um peso médio de', round(media,2), 'kg.')

# Cálculo do desvio padrão
desvio = 0
for peso in Limite_peso:
    desvio = desvio + (peso - media)**2

desvio = (desvio/len(Limite_peso))**0.5

print('E um desvio padrão de', round(desvio, 2), 'kg.')
````

## 3. Metereologia

_(Ficha 6, Problema 4)_

O ficheiro `meteo.csv` (que deve colocar na diretoria onde está a trabalhar) contem dados meteorológicos (luz e temperatura) do mês Março de uma cidade na Hungria. O ficheiro apresenta as médias de 15 em 15 minutos.

Escreva um programa que determine a data e hora em que se
verificou a temperatura mínima e máxima neste intervalo de dados.

````python
#@title Solução

f = open('meteo.csv', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

f.close()

# iniciar as variáveis necessárias
data_max = ''   # data da temperatura máxima
temp_max = 0    # valor inicial para a temperatura máxima
data_min = ''   # data da temperatura mínima
temp_min = 100  # valor inicial para a temperatura mínima

for line in linhas[1:]:         # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)

    line = line.strip('\n')
    linha = line.split(';')

    # obter as variáveis necessárias
    data = linha[0]
    temp = float(linha[2])

    # atualizar os valores
    if temp > temp_max:
        temp_max = temp
        data_max = data
    elif temp < temp_min:
        temp_min = temp
        data_min = data

# Imprimir os resultados
print('A temperatura mínima é', temp_min, 'ºC e foi obtida em', data_min)
print('A temperatura máxima é', temp_max, 'ºC e foi obtida em', data_max)
````

$$T(n) = n \times \frac{n+1}{2}$$ para $n ≥ 1$.

````python
#@title Solução

t = int(input('Introduza um número: '))

if t <= 0:
    T = 0
    n = 0
else:
    for n in range(1, t+1): # para cada n de 1 a t:
        T = n*(n+1)/2       # calcular o número triangular de ordem n
        if T == t:          # se o número triangular for igual a T, estamos na ordem certa
            break
        elif T > t:         # se o número triangular ultrapassar t, temos que voltar à ordem e ao número triangular anterior
            n = n - 1
            T = n*(n+1)/2
            break

print('O maior número triangular inferior ou igual a', t, 'é', T, 'de ordem', n)
````

## 6. Letras Maiúsculas

_(Ficha 6, Problema 8)_

Escreva um programa que leia um ficheiro de texto com o nome `texto.txt` e crie um ficheiro `texto_c.txt` onde todas as letras que se seguem a um ponto final (`.`) são convertidas na maiúscula respetiva. O programa deve ainda indicar o número total de ocorrências.
- **Nota:** comece por criar o ficheiro `texto.txt` por exemplo no `Notepad` (Bloco de notas).

````python
#@title Solução

f = open('texto.txt', mode = 'r')

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

f.close()

linhas = ''.join(linhas) # transformar todo o texto numar única string

novo_texto = linhas[0].upper()  # colocar a primeira letra em maiúscula

i = 1

while i < len(linhas)-2: # para cada posição do texto
    if linhas[i] == '.':    # quando temos um ponto, queremos copiar o ponto, o espaço seguinte e a letra em maiúscula
        novo_texto = novo_texto + linhas[i:i+2] + linhas[i+2].upper()
        i = i + 3
    else:     # quando não temos um ponto, queremos copiar o caracteres tal como é
        novo_texto = novo_texto + linhas[i]
        i = i + 1

# escrever o novo texto no ficheiro
output = open('texto_c.txt', mode = 'w')

output.write(novo_texto)

output.close()
````

## 7. Números complexos

_(Ficha 6, Problema 9)_

Escreva um programa que, a partir do ficheiro `num.txt` (que pode criar, por exemplo, no `Notepad`) do tipo:

```
Real / Imaginario
3.0 / 2.5
1.7 / 2.9
2.4 / 3.2
...
```

Obtenha um novo ficheiro designado por `complexos.txt`
do tipo:

```
N. Complexos
3.0 + 2.5i

...
```

````python
#@title Solução

f = open('num.txt', mode = 'r')             # abrir o ficheiro de leitura
output = open('complexos.txt', mode = 'w')  # criar o ficheiro de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

output.write('N. Complexos \n')     # escrever o cabeçalho

for linha in linhas[1:]:    # para cada linha obter a parte real (r) e a parte imaginária (i)
    r, i = linha.split('/')
    # L = linha.split('/')
    i = float(i.strip('\n'))
    r = float(r)
    escrever = str(r)+' + '+str(i)+'i \n'     # compor a linha que pretendemos escrever
    output.write(escrever)                  # escrever no ficheiro de saída

f.close()
output.close()
````

## 8. Linha de Montagem

_(Ficha 6, Problema 10)_

Para determinar a eficiência de produção de uma linha de montagem numa fábrica foi instalado um sistema de medição do número de peças montadas que gera um ficheiro diário, `montagem.txt`, com um registo das peças montadas de 5 em 5 minutos, iniciando a contagem às 08.00 e terminando às 19.00.

Cada registo (linha) tem dois valores: o instante em que foi feita a contagem (ex: hh.mm) e o número `n` de peças montadas nesses intervalos de 5 minutos.

Depois de analisar o ficheiro, escreva um programa que permita calcular o número de peças produzidas em cada hora e a média do número de peças montadas por hora, considerando o dia todo. Esta informação deve ser colocada num ficheiro `horaria.txt`, com a informação da média horária em cada linha e colocando no final do ficheiro a média diária.

- **Exemplo:**
  ```
  8h-9h: 277
  9h-10h: ...
  ...
  19h-20h: ...
  Media diaria (n.pecas/hora): ...
  ```

````python
#@title Solução

f = open('montagem.txt', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

D = {}

for line in linhas[2:]:         # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)
    hora = int(line[:2])        # retirar a hora e o numero de peças
    n_pecas = int(line[7:])
    hora_string = str(hora)+'h-'+str(hora+1)+'h' # criar a chave para o dicionario
    if hora_string not in D:
        D[hora_string] = n_pecas
    else:
        D[hora_string] = D[hora_string] + n_pecas

# calcular a média horária
media = 0

for hora in D:
    media = media + D[hora]
media = media / len(D)

# escrever o novo texto no ficheiro
output = open('horaria.txt', mode = 'w')

for chave in D:
    linha = chave + ': ' + str(D[chave])+'\n'
    output.write(linha)

output.write('Media diaria (n.pecas/hora): '+str(round(media,2)))

output.close()
````

O ficheiro `consumo.txt` contem os consumos de energia elétrica, em kWh, de um pequeno escritório medidos de 15 em 15 minutos para o mês de Setembro.



### a) Consumo mensal

Escreva um programa para processar este ficheiro que calcule o consumo total do mês;

````python
#@title Solução

f = open('consumo.txt', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

f.close()

Datas = []      # Lista vazia para colocar as datas
Consumo = []    # Lista vazia para colocar os consumos

for line in linhas[1:]:         # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)
    line = line.strip('\n')     # retiramos os dados que precisamos
    linha = line.split(';')
    Datas.append(linha[0])
    Consumo.append(float(linha[1]))

consumo_total = 0

for c in Consumo:
    consumo_total = consumo_total + c

print('O consumo total do mês é', round(consumo_total, 2), 'kWh.')
````

### b) Consumo diário

Escreva um programa que calcule os consumos diários totais, a média diária e determine os dias de maior e menor consumo.

````python
#@title Solução

consumo_diario = {} # dicionário vazio onde vamos colocar os dias e o seu valor do consumo

for i in range(len(Datas)): # para cada posição i das listas:
    dia = Datas[i][:10]     # retiramos o dia
    if dia not in consumo_diario:       # se for um dia novo, começamos com o consumo da primeira entrada
        consumo_diario[dia] = Consumo[i]
    else:                               # se esse dia já estiver registado, somamos o novo consumo
        consumo_diario[dia] = consumo_diario[dia] + Consumo[i]

print('Os consumos diários totais são:', consumo_diario)

# Cálculo da média diária
media_diaria = 0

for consumo in consumo_diario.values():     # somamos cada consumo diário do dicionário
    media_diaria = media_diaria + consumo

media_diaria = media_diaria / len(consumo_diario)

print('A média diária é', round(media_diaria, 2), 'kWh.')

# Dia de maior e menor consumo:

# determinar o maior e o menor consumo
maior_c = max(consumo_diario.values())
menor_c = min(consumo_diario.values())

for dia in consumo_diario:              # percorrer o dicionário até encontrar os dias correspondentes ao maior e menor consumo
    if consumo_diario[dia] == maior_c:
        print('O dia de maior consumo diário foi', dia, ':', maior_c, 'kWh.')
    elif consumo_diario[dia] == menor_c:
        print('O dia de menor consumo diário foi', dia, ':', menor_c, 'kWh.')
````

### c) Consumo semanal

Escreva um programa que determine o consumo médio diário de cada dia da semana (2ª a domingo), indicando também
a sua variação em relação à média diária.

- **Nota:** o dia 1 de setembro de 2016 foi uma 5ª feira.


````python
#@title Solução

consumo_semana = {}

mes = ['quinta', 'sexta', 'sabado', 'domingo', 'segunda', 'terça', 'quarta']*5

dia_mes = 0

# dicionário para contar quantos dias de cada tipo há
dias_semana = {'quinta': 0, 'sexta': 0, 'sabado': 0, 'domingo': 0, 'segunda': 0, 'terça': 0, 'quarta': 0}

for dia in consumo_diario:      # para cada dia:
    dia_semana = mes[dia_mes]   # determinar o dia da semana correspondente
    # somar ao dia da semana o consumo diario desse novo dia
    if dia_semana not in consumo_semana:
        consumo_semana[dia_semana] = consumo_diario[dia]
    else:
        consumo_semana[dia_semana] = consumo_diario[dia] + consumo_semana[dia_semana]
    dias_semana[dia_semana] += 1 # contar o dia da semana
    dia_mes = dia_mes + 1 # passar para o próximo dia do mês

print(dias_semana)

print('O consumo total por dia da semana é', consumo_semana)

for dia in consumo_semana: # para obter o consumo médio, temos que dividir o consumo total pelo número de dias
    consumo_semana[dia] = consumo_semana[dia] / dias_semana[dia]

print('O consumo médio por dia da semana é', consumo_semana)
````

### d) Consumo residual

Escreva um programa que calcule o consumo "residual", ou seja, o consumo médio entre as 2:00 e as 6:00 da manhã.

````python
#@title Solução

consumo_residual = {}

for i in range(len(Datas)): # para cada posição i das listas:
    dia = Datas[i][:10]     # retiramos o dia
    hora = int(Datas[i][11:13])    # e a hora
    if hora >= 2 and hora < 6:      # verificamos se a hora é do período residual e se sim, adicionamos esse consumo
        if dia not in consumo_residual:
            consumo_residual[dia] = Consumo[i]
        else:
            consumo_residual[dia] = Consumo[i] + consumo_residual[dia]

print(consumo_residual)

residual_medio = 0
for consumo in consumo_residual.values():
    residual_medio += consumo

residual_medio = residual_medio / len(consumo_residual)

print('O consumo residual médio é,', round(residual_medio, 2), 'kWh.')
````
