## Sintaxe

Tal como o ciclo `while`, o ciclo `for` permite-nos repetir um conjunto de instruções. No entanto, num ciclo `for` **temos que definir _a priori_ exatamente quantas iterações pretendemos realizar.** Isto significa que, ao contrário do ciclo `while`, o ciclo `for` não permite realizar ciclos infinitos ou de iterações indefinidas (ou infinitas).

A sintaxe do ciclo `for` é a seguinte:
```
for <variável> in <conjunto/sequência>: # para cada elemento do conjunto:
	 <instruções                # as instruções são seguidas iterativamente
```

## Iterar sobre objetos

Em Python a instrução `for` é utilizada exclusivamente para iterar sobre objetos. Na versão atual do Python todas as sequências, dicionários, conjuntos e ficheiros têm um iterador definido por defeito, que devolve cada um dos seus elementos da sequência.


### Percorrer uma Sequência de Números

O `Python` tem uma função que cria uma sequência de inteiros, a função `range()`.


#### Gerar inteiros entre `0` e um valor

Podemos gerar números inteiros entre `0` e `N-1` se usarmos `range(N)`:

````python
for i in range(10):
    print(i)
````

#### Gerar inteiros entre dois valores

Podemos gerar números inteiros entre `n` e `N-1` se usarmos `range(n, N)`:

````python
for i in range(3, 7):
    print(i)
````

#### Gerar inteiros com um passo

Podemos gerar números inteiros entre `n` e `N-1` com um passo `k` se usarmos `range(n, N, k)`:

````python
for i in range(1, 6, 2):
	print(i)
````

### Percorrer uma `string`

Podemos percorrer as letras `string` (ou de qualquer outra sequência), elemento a elemento através de um ciclo `for`:

````python
cidade = 'Coimbra'

for letra in cidade:
	print(letra)
````

### Percorrer uma `string` pelos seus índices

No entanto, com este formato **perdemos acesso aos índices de cada elemento**, o que pode ser útil em alguns casos. Nestes casos, podemos fazer:

````python
cidade = 'Coimbra'

for i in range(0, len(cidade)):
  print(i, cidade[i])
````

## Cláusulas `else`, `break` e `continue`

### Cláusula `continue`

Quando o programa chega à cláusula `continue`, o ciclo `for` passa para a iteração seguinte, sem executar as instruções que faltavam dessa iteração.

- **Exemplo:** Um programa que conte as letras de uma frase, sem contar espaços.

<details>
  <summary><font size="4">Solução</font></summary>

```
s = 'Hello world!'

count = 0    # contador de letras

for letra in s:      # para cada letra na fras
    if letra ==' ':  # se ela for um espaço:
        continue # passar à letra seguinte
    count += 1   # caso contrário, adicionar um ao contador

print('Há', count, 'letras na frase.')
```
</details>


````python
frase = 'Hello w rld!'

contador = 0

for letra in frase:

  if letra ==' ':
    continue

  contador = contador + 1

print(contador)
````

### Cláusula `break`

Quando o programa chega à cláusula `break`, o ciclo `for` termina, passando para a instrução que se segue no programa.

- **Exemplo:** Um programa que imprime a posição da primeira letra `a` que aparece numa `string`:

<details>
  <summary><font size="4">Solução</font></summary>

```
s = 'O rato roeu a rolha da garrafa do rei da Rússia.'

print(s)

for i in range(0, len(s)):
  if s[i] == 'a':
    print('A primeira letra "a" está na posição', i)
    break
```
</details>


````python
frase = 'O rato roeu a rolha da garrafa do rei da Rússia.'

contador = 0

for letra in frase:

  if letra =='a':
    break

  contador = contador + 1

print(contador)
````

### Cláusula `else`

As instruções que se seguem à cláusula `else` são executadas apenas se o ciclo `for` for executado até ao fim, isto é, quando o ciclo `for` chegar à última iteração.

Se a instrução `break` for accionada, as instruções no bloco `else` não são executadas.

- **Exemplo:** Um programa que imprime a posição da primeira letra `a` que aparece numa `string`, e que devolva uma resposta caso a frase introduzida não contenha a letra `a`:

<details>
  <summary><font size="4">Solução</font></summary>

```
s = 'O porco roeu o topo do copo do rei do Kosovo.'

print(s)

for i in range(0, len(s)):
  if s[i] == 'a':
    print('A primeira letra "a" está na posição', i)
    break

else:
  print('A frase não contém a letra "a".')
```
</details>


````python
#frase = 'O porco roeu o topo do copo do rei do Kosovo.'
frase = 'O rato roeu a rolha da garrafa do rei da Rússia.'

contador = 0

for letra in frase:

  if letra =='a':
    break

  contador = contador + 1

else:
  print('A frase não tem letra a.')

print(contador)
````

## Quando usar `while` ou `for`?

- `while`: para casos em que **não sabemos quantas repetições vão ser feitas**:
  
  - Ciclos infinitos
  - Continuar a pedir um valor até que ele seja válido
  - Continuar a jogar enquanto não houver um vencedor
  - ...

- `for`: para os casos em que **sabemos exatamente quantas iterações são necessárias**:

  - Iterar sobre objetos (percorrer uma `string`)
  - Gerar uma sequência de números
  - ...

## 2. Somatório

Escreva um programa que peça um número inteiro $n$ e calcule o resultado de:

$$\sum_{i=0}^n\frac{1}{i!}$$

````python
#@title Solução

n = int(input('Introduza um número inteiro: '))

def fatorial(n):
  fatorial = 1
  for i in range(1, n + 1):
    fatorial = fatorial * i
  return fatorial

soma = 0

for i in range(0, n+1):
  soma = soma + (1 / fatorial(i))

print('O resultado é', soma)
````

## 3. Divisores

_(Ficha 2, Problema 1)_

Escreva um programa que peça um número inteiro positivo `n` e que apresente todos os números `k` entre `1` e `n-1` pelos quais `n` for divisível. Se número introduzido não for positivo, o programa deve voltar a pedir `n` até que seja introduzido um número positivo.


````python
#@title Solução

n = -1 # iniciar o N

# testar se o input é positivo:

while n <= 0:
    n = int(input('Introduza um numero positivo: '))

print(n, ' é divisível por:') # imprimir o cabeçalho:

for k in range(1, n):   # percorrer todos os números k entre 1 e n-1
    if n % k == 0:      # testar se n é divisível por k
        print(k)
````

## 4. Produto de números ímpares

_(Ficha 2, Problema 2)_

Escreva um programa que peça um número inteiro positivo $m$ e determine o valor máximo de $n$ para o qual o produto dos números ímpares inferiores ou iguais a $n$ é inferior ou igual a $m$:

$$1 \times 3 \times 5 \times 7 \times … \times n\leq m$$

````python
#@title Solução

m = -1 # iniciar o N

# testar se o input é positivo:

while m <= 0:
    m = int(input('Introduza um numero positivo: '))

prod = 1 # inicializar o produto

for n in range (1, m+1, 2): # percorrer os numeros impares (de 2 em 2) de 1 até m
    prod = prod * n       # calcula o novo termo do produto
    if prod == m:         # se o produto ficar igual a m, estamos no n máximo
        print('O valor máximo de n é', n)
        break
    elif prod > m:        # se o produto ultrapassar m, o n máximo era o número ímpar anterior
        print('O valor máximo de n é', n - 2)
        break
````

## 6. Palíndromo

_(Ficha 3, Problema 1)_

Escreva um programa que verifique se uma palavra introduzida pelo utilizador é um palíndromo, isto é, se é igual quando lida da esquerda para a direita ou da direita para a esquerda (ex. `ana` e `osso` são palíndromos).

````python
#@title Solução

palavra = input('Introduza a palavra: ')

for i in range(len(palavra)):              # Para cada posição da palavra:
    if palavra[i] != palavra[-i-1]:     # Se o elemento com indexação positiva for igual ao de indexação negativa equivalente:
        print('A palavra não é um palíndromo.')     # A palavra não é um palíndromo e acaba o ciclo
        break
else: print ('A palavra é um palíndromo.')   # Se o ciclo chegar ao fim, então é um palíndromo
````

## 7. Expansão de Taylor

_(Ficha 5, Problema 1)_

Execute um programa que permita calcular o $\cos(x)$, para um dado valor de $x$ introduzido pelo utilizador, usando a expansão de Taylor:

$$\cos(x)=\sum_{n=0}^\infty \frac{(-1)^n}{(2n)!} \times x^{2n}$$

Este valor deve ser calculado até que erro seja menor do que um valor, `e`, introduzido pelo utilizador. Considere que esse valor foi atingido quando a diferença entre dois valores consecutivos da série calculada forem menores do que o valor `e`.

O programa deve também indicar quantas parcelas do somatório foram necessárias para calcular o valor do cosseno com a precisão exigida.

````python
#@title Solução

pi = 3.141592653589793238462643383279502884197169399375105820974944592307816406286

e = float(input('Introduza o erro pretendido: '))
teta = float(input('Introduza o ângulo em graus: '))
x = teta * pi / 180    # converte o ângulo para radianos para poder fazer as contas

n = 0                   # ordem do somatório
ultima_parcela = 100      # inicializar última parcela
cos_estimado = 0        # inicializar o cosseno estimado
erro = e + 1            # inicializar o erro calculado com um valor que entre no ciclo

while erro > e:

    fatorial = 1
    for i in range(1, 2*n+1):
      fatorial = fatorial * i

                                                           # enquanto o erro calculado for superior ao erro pretendido:
    cos_estimado = cos_estimado + ((-1)**n*x**(2*n))/(fatorial)    # adicionar a nova parcela
    erro = abs(cos_estimado - ultima_parcela)                           # calcular o novo erro
    n += 1                                                              # atualizar o iterador
    ultima_parcela = cos_estimado                                       # atualizar o valor da ultima parcela estimada

# imprimir os resultados
print(f'cos({teta}º) =', round(cos_estimado,4))
print('Foram precisas', n, 'iterações.')
````

## 8. Árvore de Números

_(Ficha 5, Problema 3)_

Escreva um programa que dado um inteiro `n` entre `0` e `9`, escreva a seguinte árvore no ecrã ([triângulo de Pascal](https://pt.wikipedia.org/wiki/Tri%C3%A2ngulo_de_Pascal)):

Exemplo para `n=5`:

```
     0
    010
   01210
  0123210
 012343210
01234543210
```

````python
#@title Solução

n = int(input('Introduza o número n: '))

# verificar que o número dado é válido:
while n < 1 or n > 9:
    n = int(input('Introduza o número n: '))

for linha in range(n + 1):          # para cada linha da árvore:
    espacos = (n + 1) - linha       # calcular quantos espaços são precisos para alinhar a linha na árvore
    print(' '*espacos, end='')      # imprimir os espaços
    for crescente in range(linha + 1):  # para a parte crescente da linha:
        print(crescente, end='')        # imprimir os números de 0 ao numero central
    while crescente > 0:                # para a parte decrescente da linha, ir tirando um ao numero anterior até chegar a 0
        crescente = crescente - 1
        print(crescente, end='')        # imprimir cada um dos números
    print('')                           # fazer um print vazio para fazer enter
````
