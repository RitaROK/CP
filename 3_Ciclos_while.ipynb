## Sintaxe

Os ciclos `while` permitem-nos executar o mesmo conjunto de instruções várias vezes, **enquanto uma determinada condição for verdadeira (`True`).**

A sintaxe de um bloco `while` tem a seguinte sintaxe:


```
while <condição>:     # enquanto a condição for verdadeira:
	<instruções>        # o bloco de instruções é executado ciclicamente
```

Tal como na seleção `if`, a **identação é essencial** para definir o conjunto de instruções que são repetidas no programa.

Cada repetição deste bloco de instruções tem o nome de _**iteração**_.


````python
while True:         # ciclo infinito
  print('Hello')
````

```
n = int(input('Introduza o valor: '))
fatorial = 1

while n < 0:

if n == 0:
	print('Fatorial de', n, '= 1')

    i = n

        fatorial = fatorial * i  # calcular a nova parcela do fatorial
```

### Cláusula `continue`

Quando o programa chega a uma cláusula `continue`, o ciclo `while` passa para a iteração seguinte, sem executar as instruções que faltavam dessa iteração.

- **Exemplo:** Um programa que conte as letras de uma frase, sem contar espaços.

<details>
  <summary><font size="4">Solução</font></summary>

```
s = 'Hello world!'

count = 0    # contador de letras
i = 0        # iterador

while i < len(s):      # para cada letra na frase

    if s[i] ==' ':  # se ela for um espaço:
        i = i + 1
        continue # passar à letra seguinte

    count += 1   # caso contrário, adicionar um ao contador
    i = i + 1

print('Há', count, 'letras na frase.')
```
</details>



````python
frase = 'Hello world!'

letra = 0

contador = 0

while letra < len(frase):

  if frase[letra] == ' ':
    letra = letra + 1
    continue

  letra = letra + 1

  contador = contador + 1

print(contador)
````

### Cláusula `break`

Quando o programa chega a uma cláusula `break`, o ciclo `while` termina, passando para a instrução que se segue no programa.

- **Exemplo:** Um programa que imprime a posição da primeira letra `a` que aparece numa frase:

<details>
  <summary><font size="4">Solução</font></summary>

```
s = 'O rato roeu a rolha da garrafa do rei da Rússia.'

print(s)

i = 0 # contador que vai percorrer as letras da frase

while i < len(s):
  if s[i] == 'a':
    print('A primeira letra "a" está na posição', i)
    break
  i = i + 1
```
</details>

````python
frase = 'O ato roeu a rolha da garrafa do rei da Rússia.'

letra = 0

contador = 0

while letra < len(frase):

  if frase[letra] == 'a':
    contador = contador + 1
    letra = letra + 1
    break

  letra = letra + 1

  contador = contador + 1

print(contador)
````

### Cláusula `else`

As instruções que contidas numa cláusula `else` são executadas apenas se o ciclo `while` for executado até ao fim, isto é, quando a condição do `while` deixar de se verificar.

Se a instrução `break` for accionada, as instruções no bloco `else` não são executadas.

- **Exemplo:** Um programa que imprime a posição da primeira letra `a` que aparece numa frase, e que devolva uma resposta caso a frase introduzida não contenha a letra `a`:

<details>
  <summary><font size="4">Solução</font></summary>

```
s = 'O porco roeu o topo do copo do rei do Kosovo.'

print(s)

i = 0 # contador que vai percorrer as letras da frase

while i < len(s):
  if s[i] == 'a':
    print('A primeira letra "a" está na posição', i)
    break
  i = i + 1

else:
  print('A frase não contém a letra "a".')
```
</details>


````python
frase = 'O rato roeu a rolha da garrafa do rei da Rússia.'

letra = 0

contador = 0

while letra < len(frase):

  if frase[letra] == 'a':
    contador = contador + 1
    letra = letra + 1
    break

  letra = letra + 1

  contador = contador + 1

else:
  print('A frase não tem letra a.')

print(contador)
````

Escreva programas que calculem a soma das seguintes séries e verifique que à medida que $N$ aumenta, a série se aproxima do valor esperado.


$$\sum_{n=0}^N a^n$$

, que converge para $1/(a-1)$ quando $|a|<1$ e $N \longrightarrow +\infty$.

````python
#@title a) Solução

a = float(input('Introduza o valor de a:'))
N = int(input('Qual o maior termo da série (N)?'))
soma = 0
n = 0

while n <= N:
    soma += a**n
    n = n+1

print(soma)
````

$$\sum_{n=0}^N n$$ , que toma o valor de $\frac{N(N+1)}{2}$.

````python
#@title b) Solução
N = int(input('Qual o maior termo da serie (N)? '))

soma = 0
n = 0

while n <= N:
    soma += n      # o somatorio=termo anterior + n
    n += 1                #atualiza o termo seguinte

print(soma)
````

$$\sum_{n=0}^N \frac{x^n}{n!}$$ , que tende para $e^x$ quando $N \longrightarrow +\infty$.

<details>
  <summary><font size="4">Algoritmo</font></summary>

```
Igual ao 3.1)
```

````python
#@title c) Solução

N = int(input('Qual o maior termo da serie? '))
x = int(input('Qual o valor de x na serie? '))

soma = 0
n = 0

while n <= N:

    fatorial = 1 # inicializa a variável fatorial
    i = n        # O que se segue é o programa 1. para calcular o fatorial dentro de cada parcela
    while i > 1:
        fatorial = fatorial * i
        i = i- 1

    soma += x**n / fatorial      # atualizar o valor de soma
    n += 1                       # atualiza o termo seguinte

print(soma)
````

$$\sum_{n=0}^N \frac{(-1)^n \times x^{2n+1}}{(2n+1)!}$$, que tende para $\sin(x)$ quando $N \longrightarrow +\infty$.

<details>
  <summary><font size="4">Algoritmo</font></summary>

```
Igual ao 3.3)
```

````python
#@title d) Solução 1

N = int(input('Qual o maior termo da serie? '))
x = int(input('Qual o valor de x na serie? '))

soma = 0
n = 0

while n <= N:

    fatorial = 1 # inicializa a variável fatorial
    i = 2*n+1        # O que se segue é o programa 1. para calcular o fatorial dentro de cada parcela
    while i > 1:
        fatorial = fatorial * i
        i = i- 1

    soma += ((-1)**n * x**(2*n+1)) / fatorial     # atualizar o valor de soma
    n += 1                       # atualiza o termo seguinte

print(soma)
````

````python
#@title d) Solução 2 (mais eficiente, mas mais complicada)

N = int(input('Qual o maior termo da serie? '))
x = int(input('Qual o valor de x na serie? '))

fatorial = 1       # fatorial = (2n+1)!: o primeiro termo é 1
soma = 0
n = 0

while n <= N:
    soma += ((-1)**n * x**(2*n+1)) / fatorial
    n += 1
    fatorial = fatorial * (2*n) * (2*n+1)   # para calcular o fatorial que precisamos a seguir, que é dois termos à frente!

print(soma)
````

## 6. Série de Fibonacci

_(Ficha 1, Problema 2)_

A [sequência de Fibonacci](https://pt.wikipedia.org/wiki/Sequ%C3%AAncia_de_Fibonacci) é uma série de números inteiros onde o primeiro termo é `0` e o segundo termo é `1`. A partir do terceiro termo, cada termo é a soma dos dois anteriores.

Escreva um programa para escrever os primeiros $N$ números da sequência de Fibonacci.


<details>
  <summary><font size="4">Algoritmo</font></summary>

```
Pede um número máximo de elementos N
INICIALIZA o contador, i, e os dois primeiros elementos da sequência, F1 e F2.
IMPRIME os dois primeiros elementos da sequência
ENQUANTO o contador for menor que N:
  Calcula e imprime o próximo elemento da sequência, F3
  Atualiza dos últimos dois elementos da sequeência e o contador
```
</details>

````python
#@title Solução

N = int(input('Introduza o valor de N: '))

F1 = 0     # 1º elemento
F2 = 1     # 2º elemento
i = 2      # contador

print(F1, F2, end=' ')   # imprime os dois primeiros elementos da sequência

while i < N:
    F3 = F1 + F2
    print(F3, end = ' ')
    F1 = F2
    F2 = F3
    i += 1     # soma uma unidade ao contador / é igual a (i = i+1)
````

### a) Verificação

Escreva um programa que verifique se um dado número é primo.

<details>
  <summary><font size="4">Algoritmo</font></summary>

```
PEDE um número inteiro n
INICIA um contador em i=2 (primeiro divisor)
ENQUANTO i < n:
  Verifica SE o número n é divisível por i:
      - True -> Não é primo e acaba o programa
  Soma uma unidade ao contador
SENÃO for divisível por nenhum número, é porque é primo
```
</details>

````python
#@title Solução

n = int(input('Introduza n: '))

i = 2 # primeiro divisor a testar

while i < n:
    if n % i == 0:
        print (n, 'não é primo.')
        break
    i = i + 1

else:
    print (n, 'é primo.') #se nunca ativar o break, n e primo e imprime no ecra
````

## 9. Série $\pi$

_(Ficha 1, Problema 6)_

Uma série convergente para o cálculo de $\pi/2$ tem a seguinte forma:

$$\frac{\pi}{2} \approx p(n) =\frac{2 \times 2}{1 \times 3} \times \frac{4 \times 4}{3 \times 5} \times \frac{6 \times 6}{5 \times 7} … \frac{ 2n \times 2n }{(2n-1) \times (2n +1)}$$

Escreva um programa que calcule $n$ de modo que:

$$\bigg|\frac{\pi}{2}-p(n) \bigg| < 0.0001$$

**- Nota:** Para determinar o módulo $|x|$, pode utilizar a função `abs(x)`.

<details>
  <summary><font size="4">Algoritmo</font></summary>

```
ENQUANTO a diferença (delta) do resultado for superior a 0.0001:
  CALCULA a próxima parcela e o novo desvio
No fim, imprime o valor da série, a desvio atingido e o número de iterações necessárias
```
</details>

````python
#@title Solução
pi = 3.141592653589793238462643383279502884197169399375105820974944592307816406286

n = 1 # inicia os fator n
prod = 1 # inicia o produto
delta = 1 # diferença inicial

while delta>0.0001:
    prod = prod * (2*n * 2*n) / ((2*n-1)*(2*n+1)) # calcular a nova parcela
    delta = abs(pi/2 - prod) # calcular o novo desvio
    n = n+1  # obtém o index da próxima parcela

print (f'Valor calculado de pi/2:', prod, '\nErro:', delta,'\nIterações:', n-1)
````

## 10. Múltiplos com `1`

_(Ficha 3, Problema 2)_

Todos os **números ímpares não divisíveis por `5`** possuem pelo menos um múltiplo (diferente da unidade) que se escreve apenas com o algarismo `1`. Por exemplo:
- `111` é múltiplo de `3`
- `111111` é múltiplo de `7`

Escrever um programa que determine o menor múltiplo de um número (a pedir ao utilizador) que se escreve apenas com `1`. O programa deve rejeitar os números que não satisfaçam as condições referidas.

````python
#@title Solução

N = 5

while N % 2 == 0 or N % 5 ==0:          # Enquanto o número for par ou divisível por 5
    N = int(input('Escreva um número N ímpar não divisível por 5: '))     # continua a pedir um número ao utilizador

multiplo = '1' # o primeiro multiplo a ser avaliado

while int(multiplo) % N != 0:  # enquanto não encontrarmos o múltiplo de N:
    multiplo = multiplo + '1'  # acrescentar um novo 1 ao múltiplo

print('O menor múltiplo de', N, 'que apenas contenha "1" é', multiplo, 'e tem', len(multiplo), 'algarismos.')
````
