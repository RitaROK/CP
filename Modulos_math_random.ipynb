## Módulo `random`

Para utilizar números aleatórios em `Python`, podemos usar o módulo de números aleatórios `random`.



### Gerar números aleatórios entre 0 e 1.

Com a função `random()` podemos gerar números (pseudo-)aleatórios no intervalo [0,1).

````python
import random

r = random.random()
print(r)
````

### A _seed_

Podemos sempre gerar os mesmos números aleatórios, o que pode ser útil para testar alguns programas:

````python
from random import random, seed

seed(1)  # inicializa o gerador de números aleatórios sempre na posição 1

print(random(), random(), random())
````

### Gerar aleatórios entre dois números

Os aleatórios gerados com a função `random()` são valores entre `0` e `1`, mas este intervalo pode ser facilmente alterado para um valor entre `(min, max)`:

````python
from random import random

m = -3
M = 3

scaled_random = m + (random() * (M-m))  # gera um numero aleatorio entre m e M
print(scaled_random)
````

### Gerar números aleatórios inteiros

Podemos também gerar valores aleatórios inteiros, usando a função `randint(m, n)`, que gera um aleatório inteiro no intervalo `[m,n]`, que inclui ambos os valores:

````python
from random import randint

print(randint(2, 89))
````

### Extrair uma amostra aleatória

A função `sample(seq, k)` extrai aleatoriamente uma lista com `k` elementos de uma sequência `seq`.

Por exemplo, podemos sortear 5 letras aletórias do alfabeto:

````python
from random import sample

alfabeto = 'ABCDEFGHIJKLMNOPQRSTUVXZ'

print(sample(alfabeto, 5))
````

### A importância de Gerar Números Aleatórios

Gerar números verdadeiramente aleatórios é extremamente importante para encriptar dados digitais, para assegurar que um _hacker_ é incapaz de descobrir a chave e aceder aos nossos dados pessoais. Como tal, os dados não podem ser encriptados a partir destes números "aleatórios".

Para gerar números verdadeiramente aleatórios, os computadores baseam-se em fenómenos naturais do mundo real que sejam verdadeiramente imprevisíveis e aleatórios. Como exemplo, a Cloudfare usa cerca de 100 lâmpadas de lava que são fotografadas periodicamente de modo a gerar uma `string` de números verdadeiramente aleatórios e gerar chaves de encriptação.

## 1. Lançamento de dados

### a) Simulação de 100 lançamentos

_(Ficha 3, Problema 3)_

Escreva um programa que simule o lançamento de dois dados 100 vezes. O programa deve escrever, para cada lançamento, os valores dos dois dados (`1` até `6`) e a soma dos pontos.


````python
#@title Solução

from random import randint, seed

seed(1)

for i in range(100):   # Para cada lancamento, i, em 100:
    print('Para o lançamento', i+1, ':')
    dado_1 = randint(1,6)   # o dado 1 tem um valor aleatório entre 1 e 6
    print('O dado 1 obteve', dado_1)
    dado_2 = randint(1,6)   # a mesma coisa com o dado 2
    print('O dado 2 obteve', dado_2)
    print('A soma dos pontos é', dado_1 + dado_2)
````

### b) Histograma de lançamentos

_(Ficha 3, Problema 4)_

Modifique o programa anterior de forma a criar um programa que, utilizando um dicionário, conte o número de vezes que, no total dos 100 lançamentos, a soma é igual a `2`, `3`, ..., `12`.

````python
#@title Solução

from random import randint, seed

seed(1)

D = {}          # inicializar o dicionário vazio

for i in range(100):   # Para cada lancamento, i, em 100:
    dado_1 = randint(1,6)   # o dado 1 tem um valor aleatório entre 1 e 6
    dado_2 = randint(1,6)   # a mesma coisa com o dado 2
    soma = dado_1 + dado_2  # calcula a soma entre os dois dados

    if soma not in D.keys():    # se o resultado da soma ainda não estiver no dicionário
        D[soma] = 1             # é a primeira vez que lancamos essa pontuação
    else:                       # senão:
        D[soma] = D[soma] + 1   # temos que adicionar uma unidade a esse valor do dicionário

# Vamos imprimir os resultados finais:
for i in D.keys():
    print('A pontuação', i, 'foi obtida', D[i], 'vezes.')
````

Faça um programa que gere uma lista valores de 15 valores inteiros aleatórios entre `-100` e `100` inclusive, e que imprima a lista no ecrã.

````python
#@title Solução

from random import randint, seed

seed(1)

Lista = []                              # Inicializa uma lista vazia

for i in range(15):                     # Para cada um dos 15 elementos:
    Lista.append(randint(-100,100))     # Adiciona um número aleatório entre -100 e 100

print('A lista é', Lista)
````

### b) Identificar números negativos

Complete o programa de modo a escolher os números da lista anterior que são negativos, escrevendo esses valores negativos no ecrã, separados por um espaço em branco. Caso a lista não tenha valores negativos o programa deve escrever no ecrã: `Não existem números negativos na lista.`


````python
#@title Solução

from random import randint, seed

seed(1)

Lista = []

for i in range(15):
    Lista.append(randint(-100,100))

print('A lista é', Lista)

neg = 0 # contador de numeros negativos

for i in Lista:     # Para cada elemento da lista:
    if i < 0:       # Se ele for negativo
        if neg == 0:    # Se ele for o primeiro negativo, imprime uma frase introdutória:
            print('Os números negativos da lista são: ', end='')
        print(i, end=" ") # Imprime o número negativo
        neg = neg + 1     # acrescenta uma unidade ao contador

if neg == 0:  # Se não houver números negativos na lista:
    print('Não existem números negativos na lista.')
````

### c) Classificar os numeros

Crie um dicionário, onde as chaves são `neg`, `pos` e `nulo` e os valores associados são o nº de elementos da lista da alínea a) que são negativos, positivos ou iguais a zero, respetivamente. Imprima no ecrã as chaves do dicionário e os respetivos valores.


````python
#@title Solução

D = {'neg': 0, 'pos': 0, 'nulo': 0}  # inicializar o dicionário

Lista = []

for i in range(15):
    Lista.append(randint(-100,100))

print('A lista é', Lista)

for i in Lista:             # Para cada elemento da lista
    if i == 0:              # Se for igual a 0, acrescenta 1 ao nulo
        D['nulo'] += 1
    elif i > 0:             # Se for maior que 1, acrescenta 1 ao pos
        D['pos'] += 1
    else:                   # Senão, é negativo e acrescenta 1 ao neg
        D['neg'] += 1

for i in D.keys():                     # Para cada chave no dicionário
    print('Há', D[i], end=' ')      # Imprime o valor e pela chave imprime se é:
    if i == 'pos':                   # positivo
        print('positivo', end='')
    elif i == 'neg':                # negativo
        print('negativo', end='')
    else:
        print('nulo', end='')       # nulo
    if D[i] != 1:                   # se o valor for maior que um, acrescenta um s para fazer o plural
        print('s')
````

## 3. Palavra chave

_(Ficha 5, Problema 5)_

Escreva um programa que origine uma palavra-chave (`string`) que tenha como carateres possíveis as letras do alfabeto (`a`-`z` ou `A`-`Z`) e algarismos (`0`-`9`) e que cumpra as seguintes condições:

- Ter entre 6 e 16 carateres;
- Ter pelo menos um algarismo (`0`-`9`);
- Ter pelo menos uma letra maiúscula;
- Ter pelo menos uma letra minúscula;


**Nota:** para gerar as listas de letras do alfabeto:

```
import string
Letras_minusculas = list(string.ascii_lowercase)
Letras_maisculas = list(string.ascii_uppercase)
```

````python
#@title Solução

from random import randint, sample, seed

seed(1)

n_caracteres = randint(6, 16)  # sortear o numero de caracteres

elementos = []

n_algarismos = randint(1, n_caracteres - 2) # sortear o numero de algarismos: deixar pelo menos 2 espaços para maiusculas e minusculas
for algarismo in range(n_algarismos):       # gerar os diferentes algarismos
    elementos.append(randint(0,9))

import string

n_minusculas = randint(1, n_caracteres - n_algarismos - 1) # sortear o numero de letras minusculas: deixar pelo menos um espaço para maiusculas
minusculas = sample(list(string.ascii_lowercase), n_minusculas) # sortear as letras minusculas

n_maiusculas = n_caracteres - n_algarismos - n_minusculas   # determinar o numero de letras maiusculas
maiusculas = sample(list(string.ascii_uppercase), n_maiusculas) # sortear as letras maiusculas

elementos = elementos + minusculas + maiusculas  # adicionar todos os elementos sorteados a uma lista

palavra_chave='' # inicializar a palavra-chave

# fazer shuffle dos diferentes caracteres para gerar a palavra-passe:
while len(elementos) > 0:                   # enquanto não tirarmos todos os elementos:
    indice = randint(0, len(elementos)-1)   # sortear o indice do elemento que vamos adicionar a seguir
    palavra_chave = palavra_chave + str(elementos[indice])  # adicionar esse elemento à paravra-chave
    del elementos[indice]                                   # retirar esse elemento da lista que estamos a esvaziar

print('A palavra-chave é', palavra_chave)
````

## 4. Partícula Aleatória

_(Ficha 3, Problema 8)_

Considere uma partícula que se desloca a duas dimensões. Inicialmente a partícula encontra-se na posição $(0,0)$ e, em cada iteração, desloca-se uma unidade numa direção aleatória (definida por um ângulo $[0,2\pi]$​.



### a) Distância à origem

Escreva um programa que obtenha a distância da partícula à origem $d(t)^2=x(t)^2+y(t)^2$, função do número de iterações, $t$.

````python
#@title Solução
from random import random, seed
from math import pi, cos, sin, sqrt

seed(1)

# Posição inicial:
x = 0
y = 0

t = int(input('Quantas iterações deve fazer? '))

for i in range(t):          # Para cada iteração:
    teta = random()*2*pi    # Gera o teta aleatório
    x = x + cos(teta)       # Atualiza as coordenadas x e y
    y = y + sin(teta)

d = sqrt(x**2+y**2)         # Calcula a distância final

print('A distância percorrida ao fim de', t, 'iterações é', round(d,2))
````

### b) Distância Média

 Utilizando o programa anterior, faça um outro programa que, repetindo o passo anterior para $N$ partículas ($N$ pedido ao utilizador), calcule o valor médio de $d(t)^2$ para um dado valor de $t $ (correspondente a um instante $t$ e a $t$ iterações).


````python
#@title Solução

from random import random, seed
from math import pi, cos, sin, sqrt

seed(1)

t = int(input('Quantas iterações deve fazer? '))
N = int(input('Introduza o nº de partículas: '))

# Posição inicial: agora é uma lista, com um elemento para cada partícula
x = [0] * N
y = [0] * N
d2 = [0] * N # distância percorrida por cada partícula

for n in range(N):                  # Para cada partícula:
    for i in range(t):              # Para cada iteração:
        teta = random()*2*pi        # Gera o teta aleatório
        x[n] = x[n] + cos(teta)     # Atualiza as coordenadas x e y
        y[n] = y[n] + sin(teta)
    d2[n] = x[n]**2 + y[n]**2       # Calcula a d**2 final da partícula

media = sum(d2)/N # calcula a media

print('A média para', N, 'partículas e', t, 'iterações é', round(media,2))
````

### c) Movimento browniano

Mostre que quando a média é realizada com um número elevado de simulações, se obtém, aproximadamente, que $\bar{d(t)^2} \approx K t$. Obtenha a constante $K$se aproxima à constante de difusão num movimento browniano).


````python
#@title Solução

from random import random, seed
from math import pi, cos, sin, sqrt

seed(1)

t = int(input('Quantas iterações deve fazer? '))
N = int(input('Introduza o nº de partículas: '))

# Posição inicial: agora é uma lista, com um elemento para cada partícula
x = [0] * N
y = [0] * N
d2 = [0] * N # distância percorrida por cada partícula

for n in range(N):
    for i in range(t):              # Para cada iteração:
        teta = random()*2*pi        # Gera o teta aleatório
        x[n] = x[n] + cos(teta)     # Atualiza as coordenadas x e y
        y[n] = y[n] + sin(teta)
    d2[n] = x[n]**2 + y[n]**2       # Calcula a d**2 final da partícula

media = sum(d2)/N   # calcula a media

K = media / t       # calcula a constante K

print('A constante K para', N, 'partículas e', t, 'iterações é', round(K,4))
````

O jogo desenrola-se da seguinte forma:

2. o outro jogador (chamado Python) dá um palpite para a sequência secreta (<font color="red">`R`</font><font color="green">`G`</font><font color="blue">`B`</font><font color="cyan">`C`</font><font color="magenta">`M`</font>, por exemplo);

- **Nota:** Se a sequência secreta for, por exemplo, <font color="red">`R`</font><font color="green">`G`</font><font color="blue">`B`</font><font color="cyan">`C`</font><font color="magenta">`M`</font>, e o palpite for <font color="red">`R`</font><font color="red">`R`</font><font color="yellow">`Y`</font><font color="black">`K`</font><font color="yellow">`Y`</font>, a resposta de Monty deverá ser `1` `0` `-` `-` `-`, indicando que o primeiro <font color="red">`R`</font> está na posição correta e que o segundo <font color="red">`R`</font> é uma cor presente na sequência secreta mas noutra posição.

4. Python dá um novo palpite tentando usar a informação entretanto adquirida (<font color="red">`R`</font><font color="yellow">`Y`</font><font color="black">`K`</font><font color="green">`G`</font><font color="black">`K`</font>, por exemplo) e Monty indica de novo que parte do palpite está correta (`1` `0` `-` `0` `0` neste exemplo);

5. o jogo prossegue até Python acertar na sequência secreta. No fim do jogo contam-se quantos palpites teve Python de dar até acertar e recomeça-se o jogo sendo agora Monty quem vai adivinhar uma sequência escolhida por Python. Ganha o jogo quem precisar de menos palpites.

Escreva um programa que partindo de uma sequência “secreta” de 5 caracteres com algumas das letras <font color="red">`R`</font>, <font color="green">`G`</font>, <font color="blue">`B`</font>, <font color="cyan">`C`</font>, <font color="magenta">`M`</font>, <font color="yellow">`Y`</font> e <font color="black">`K`</font> peça palpites ao utilizador até ele acertar na sequência gerada no início. Em resposta a cada palpite, o programa deve indicar que parte do palpite está correta usando o código acima descrito. No final do programa deve ser impressa a pontuação obtida pelo utilizador (número de palpites).

- **Nota:** para gerar a sequência secreta que se pretende adivinhar, pode utilizar a função `random.sample(sequencia, n)`, que gera combinações aleatórias de `n` valores, obtidos a partir da sequência.


````python
#@title Solução

from random import sample, seed

seed(1)

# Gerar o código secreto para o utilizador adivinhar
codigo_secreto = sample('RRRRRGGGGGBBBBBCCCCCMMMMMYYYYYKKKKK', 5)

pontos = 0 # inicializa o nº de palpites/pontuação
acertou = 0 # flag que passa a 1 apenas quando o utilizador acerta

while acertou == 0:                                 # enquanto o utilizador não acertar:
    palpite = input('Qual é o seu palpite? ')
    resposta = ['-', '-', '-', '-', '-']            # inicializa a resposta a dar ao utilizador como tudo errado
    for i in range(len(palpite)):                   # para cada posição do palpite:
        if palpite[i] == codigo_secreto[i]:         # se a letra nessa posição for igual à do código secreto:
            resposta[i] = '1'                       # altera o elemento correspondente da resposta para 1
        elif palpite[i] in codigo_secreto:          # Senão, se a letra estiver no código secreto:
            resposta[i] = '0'                       # altera o elemento correspondente da resposta para 0
    if resposta == ['1', '1', '1', '1', '1']:       # Se a resposta estiver certa: so 1's:
        acertou = 1                                 # Muda a flag para sair do ciclo
    else:                                           # Se não acertou:
        print('Wrong guess.')
        print(resposta)                             # Imprime a resposta
    pontos = pontos + 1                             # Adiciona 1 ao contador de pontos
else:                                               # Quando o ciclo chega ao fim:
    print('Acertou no código secreto após', pontos, 'tentativas!')  # Dá a pontuação final ao jogador
    print('Game over!')
````
