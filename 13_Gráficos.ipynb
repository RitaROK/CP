# 1. Histograma

_(Ficha 7, Problema 1)_

Usando os valores no ficheiro `analise.txt` (problema 2, da folha 6) e através da leitura dos dados deste ficheiro, represente graficamente num gráfico de barras a distribuição obtida para as somas dos 2 dados.

````python
#@title Solução

import matplotlib.pyplot as plt # importar o módulo de fazer gráficos

f = open('analise.txt', mode = 'r')    # criar o ficheiro de output no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

f.close()

# criar uma lista vazia para os dados de cada um dos eixos do gráfico
soma = []
contagem = []

for linha in linhas[1:]:        # para cada linha, adicionar a soma e a contagem correspondente à lista que lhe diz respeito
    linha = linha.strip('\n')
    dados = linha.split(',')
    soma.append(int(dados[0]))
    contagem.append(int(dados[1]))

# Fazer o gráfico de barras
plt.bar(soma, contagem)
plt.title('Soma de 10,000 lançamentos de 2 dados', fontsize = 18) # título
plt.xlabel('Soma das pontuações', fontsize = 12)    # legendas dos eixos
plt.ylabel('Frequência absoluta', fontsize = 12)
plt.tick_params(axis = 'both', labelsize = 10)   # aumentar o tamanho de letra dos eixos
plt.show()
````

## 2. Bagagem de avião

_(Ficha 7, Problema 2)_

O ficheiro `bagagem.txt` contém o peso em kg de 500 malas embarcadas num avião. Usando os valores retirados deste ficheiro, represente graficamente o peso de cada mala, num gráfico de pontos.

No mesmo gráfico represente o valor do peso médio das malas numa linha horizontal a preto e em duas linhas a verde, acima e abaixo, o valor médio $\pm$ o desvio padrão (já calculados no problema 3 da folha 6).

````python
#@title Solução

import matplotlib.pyplot as plt

# mesma resolução anterior
f = open('bagagem.txt', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

malas = [] # lista vazia com as malas acima do peso

for line in linhas[1:]:         # para cada linha do ficheiro (saltamos a primeira, que tem o cabeçalho)
    peso = float(line.strip('\n'))      # retira o peso

    malas.append(peso)

f.close()

# Cálculo da média
media  = 0
for peso in malas:
    media = media + peso

media = media / len(malas)

print ('Estas malas têm um peso médio de', round(media,2), 'kg.')

# Cálculo do desvio padrão
desvio = 0
for peso in malas:
    desvio = desvio + (peso - media)**2

desvio = (desvio/len(malas))**0.5

print('E um desvio padrão de', round(desvio, 2), 'kg.')

plt.scatter(range(len(malas)), malas)           # gráfico de dispersão: nº1 da mala versus peso
plt.axhline(y = media, color='black', ls='--', lw=2)           # valor medio
plt.axhline(y = media+desvio, color='green', ls='--', lw=2)    # media + desvio padrão
plt.axhline(y = media-desvio, color='green', ls='--', lw=2)    # média - desvio padrão
plt.title('Bagagem', fontsize = 18) # título
plt.xlabel('Nº da mala', fontsize = 12)    # legendas dos eixos
plt.ylabel('Peso da mala (kg)', fontsize = 12)
plt.tick_params(axis = 'both', labelsize = 10)   # aumentar o tamanho de letra dos eixos

plt.show()
````

## 3. Sensor de temperatura

_(Ficha 7, Problema 5)_

Pretende-se implementar um programa que leia um ficheiro de temperaturas em 3 pontos diferentes, `A`, `B` e `C`, de uma instalação fabril, para vários instantes diferentes. A leitura dos sensores em cada instante será colocada em linhas sucessivas do ficheiro.

### a) Gerar o ficheiro de dados

Comece por colocar instruções no programa que lhe permitam gerar um ficheiro que será usado para testar o programa que pretende desenvolver. Os dados a guardar neste ficheiro de teste (que pode designar por `temperatura.txt`) devem ser gerados de forma aleatória e devem variar:
- Sensor `A`: entre 20ºC - 40ºC
- Sensor `B`: entre 35ºC - 55ºC
- Sensor `C`: entre 45ºC - 65ºC.

A primeira linha do ficheiro deve conter a seguinte informação:

```
Hora - A, B, C
```

E as seguintes:

```
00:00 - 23.54, 54.65, 55.38
00:30 - ...
```

O ficheiro criado deve ter linhas entre as 00.00 e as 23.30, cada uma separada de 30 min.

- **Nota:** a função `uniform(a,b)` devolve valores aleatórios do tipo `float` entre `a` e `b`.

````python
#@title Solução

from random import uniform

output = open('temperatura.txt', mode = 'w')    # criar o ficheiro de output no modo de escrita

output.write('Hora - A, B, C\n')    # imprimir o cabeçalho

for hora in range(24):                      # para cada hora das 0 às 24:
    sensor_A = round(uniform(20, 40), 2)    # gerar as temperaturas aleatórias para os 3 sensores na hora certa
    sensor_B = round(uniform(35, 55), 2)
    sensor_C = round(uniform(45, 65), 2)
    if hora < 10:                           # se a hora for menor que 10, temos que acrescentar um 0 antes à string para compor a hora
        linha_1 = '0'+str(hora)+':00 - '
    else:                                   # senão, não é preciso
        linha_1 = str(hora)+':00 - '
    linha_1 = linha_1 + str(sensor_A) + ', ' + str(sensor_B) + ', ' + str(sensor_C) + '\n'  # compor a linha a escrever para a hora certa
    output.write(linha_1)                   # escrever no ficheiro
    sensor_A = round(uniform(20, 40), 2)    # gerar as temperaturas aleatórias para os 3 sensores para a meia hora e fazer o mesmo que o caso anterior
    sensor_B = round(uniform(35, 55), 2)
    sensor_C = round(uniform(45, 65), 2)
    if hora < 10:
        linha_2 = '0'+str(hora)+':30 - '
    else:
        linha_2 = str(hora)+':30 - '
    linha_2 = linha_2 + str(sensor_A) + ', ' + str(sensor_B) + ', ' + str(sensor_C) + '\n'
    output.write(linha_2)

output.close()
````

### b) Leitura dos dados

Coloque instruções no programa que permitam ler os dados de ficheiros do tipo do ficheiro criado (`temperatura.txt`).

Pretende-se calcular a média dos valores lidos para cada sensor e o valor máximo e mínimo também para cada um dos sensores.

Esta informação deve ser impressa no ecrã na forma:

```
Sensor A: media=32.65, max-min=15.15
Sensor B: ...
```

````python
#@title Solução

Horas = []
Sensor_A = []
Sensor_B = []
Sensor_C = []

f = open('temperatura.txt', mode='r')     # abre o ficheiro no modo de escrita

linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings

for linha in linhas[1:]:        # para cada linha
    linha = linha.strip('\n')
    linha = linha.split(',')    # separar as colunas
    time = linha[0].split('-')[0]   # separar o tempo da primeira coluna, com '-'
    hora = int(time.split(':')[0])  # retirar a hora
    minutos = int(time.split(':')[1])   # retirar os minutos
    if minutos == 30:                   # atualizar a hora se for meia hora
        hora = hora + 0.5
    Horas.append(hora)                  # adicionar os valores a cada lista
    Sensor_A.append(float(linha[0].split('-')[1]))
    Sensor_B.append(float(linha[1]))
    Sensor_C.append(float(linha[2]))

# Calcular as médias e os valores máximos e mínimos

media_A = sum(Sensor_A)/len(Sensor_A)
print('Sensor A: media =', round(media_A,2), ', max-min =', round(max(Sensor_A) - min(Sensor_A),2))

media_B = sum(Sensor_B)/len(Sensor_B)
print('Sensor B: media =', round(media_B,2), ', max-min =', round(max(Sensor_B) - min(Sensor_B),2))

media_C = sum(Sensor_C)/len(Sensor_C)
print('Sensor C: media =', round(media_C,2), ', max-min =', round(max(Sensor_C) - min(Sensor_C),2))
````

Coloque instruções num programa que lhe permitam representar graficamente os valores da temperatura dos sensores `A`, `B` e `C` em função do tempo (`t(horas)=0, 0.5, 1.0, 1.5, 2,…, 23.5`):

## 4. _Random Walk_ ou Caminhada do Bêbedo

_(Ficha 7, Problema 4)_

Considere uma partícula que se desloca a duas dimensões. Inicialmente a partícula encontra-se na posição
(`x_start`, `y_start`) e, em cada iteração, desloca-se uma unidade numa direção aleatória (definida por um ângulo $\theta$ arbitrário pertencente ao intervalo $[0 , 2\pi]$).

### a) Função _random walk_

Defina uma função `R_Walk(n, x_start, y_start)` que, a partir do número de iterações introduzido, `n`, devolva duas listas, uma lista com as posições em `x` e outra com as posições em `y`, dessa partícula.

````python
#@title Solução

from random import random
from math import pi, sin, cos

def R_walk(n, x_start, y_start):
    # adicionar a posição inicial às listas
    x = [x_start]
    y = [y_start]

    for iteracao in range(1,n):     # para cada iteração
        teta = random() * 2 * pi    # gerar o ângulo aleatório
        x.append(x[-1] + cos(teta)) # somar à última posição o novo passo
        y.append(y[-1] + sin(teta)) # somar à última posição o novo passo
    return x, y
````

### b) Trajetos aleatórios

Use a função anterior e o módulo gráfico para visualizar esses trajetos aleatórios para duas partículas. Utilize cores diferentes para cada partícula.

````python
#@title Solução

iter = int(input('Introduza o número de iterações que pretende: '))

xi_1 = float(input('Introduza o x inicial da partícula 1: '))
yi_1 = float(input('Introduza o y inicial da partícula 1: '))

x_1, y_1 = R_walk(iter, xi_1, yi_1)      # calcular as posições da partícula 1
plt.plot(x_1, y_1, color='blue', label='Partícula 1')

xi_2 = float(input('Introduza o x inicial da partícula 2: '))
yi_2 = float(input('Introduza o y inicial da partícula 2: '))

x_2, y_2 = R_walk(iter, xi_2, yi_2)      # calcular as posições da partícula 2
plt.plot(x_2, y_2, color='orange', label='Partícula 2')

plt.title('Random walk', fontsize = 18)
plt.xlabel('x position', fontsize = 12)
plt.ylabel('y position', fontsize = 12)
plt.tick_params(axis = 'both', labelsize = 8)
plt.legend()
plt.show()
````

## 5. Representar graficamente uma função

_(Ficha 7, Problema 3)_

### a) Função para gerar os valores $x$ e $y$

Defina uma função `f_graf (f, xi, xf, n)` que devolva duas listas, uma com os `n` valores de `x` desde um valor inicial, `xi`, até um valor final, `xf`, e outra lista com `n` valores da função `f(x)` para os mesmos valores de `x` (a função `f` deve ser um parâmetro de entrada na função `f_graf`).


````python
#@title Solução

def f_graf(f, xi, xf, n):
    # colocar o primeiro elemento em cada uma das listas
    lista_x = [xi]
    lista_y = [f(xi)]
    passo = abs(xf-xi)/n    # calcular o passo entre cada valor de x
    for i in range(1, n+1):            # para cada valor de x
        lista_x.append(lista_x[-1] + passo) # adicionar o novo elemento à lista: último elemento + passo
        lista_y.append(f(lista_x[-1]))      # calcular o valor de y correspondente
    return lista_x, lista_y                 # devolver as listas
````

Os valores `xi`, `xf` e `n` devem ser introduzidos pelo utilizador.
