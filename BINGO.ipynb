## a) Gerar cartões

Os cartões para jogar BINGO têm 5 colunas, cada uma com 5 números. Cada coluna têm uma letra associada:
1. `B`, para a primeira coluna;
2. `I` para a segunda;
3. `N` para a terceira;
4. `G` para a quarta;
5. `O` para a quinta.

Nas primeira colunas os números podem ser entre `1` e `15`, na segunda entre `16` e `30`, na terceira entre `31` e `45`, na quarta entre `46` e `60` e na quinta entre `61` e `75`. Os números estão ordenadas por ordem crescente em cada coluna.

Escreva um programa que permita gerar e guardar de modo apropriado um cartão de bingo.


```python
#@title Solução

def gerar_cartao():

  from random import sample

  cartao = {}

  inicio = 1
  fim = 15

  for letra in "BINGO":
    numeros = list(range(inicio, fim + 1)) # gerar os possíveis numeros
    cartao[letra] = sample(numeros, 5) # sortear 5 numeros dessa lista
    cartao[letra].sort() # ordenar os números
    inicio += 15
    fim += 15

  return cartao

cartao = gerar_cartao()

print(cartao)
```

## b) Visualizar cartões

Escreva um segundo programa que permita visualizar um cartão de bingo, imprimindo-o no ecrã.

```python
#@title Solução


def imprimir_cartao(D):
  print('-'*18)
  print('|  B  I  N  G  O |')
  print('-'*18)

  for linha in range(0, 5):
    print('|', end = '')
    for coluna in D:
      if len(str(D[coluna][linha])) < 2:
        print(f'  {D[coluna][linha]}', end = '')
      else:
        print(f' {D[coluna][linha]}', end = '')
    print(' |')

  print('-'*18)

imprimir_cartao(cartao)
```

## c) Jogo

Acrescente instruções que lhe permitam jogar com um cartão gerado, lançando números aleatórios e verificando se os números constam do cartão. Contabilize o número de jogadas necessárias para terminar o jogo, ou seja, acertar em todos os números. Em cada jogada vá substituindo o número acertado por um `-` no cartão. O jogo termina quando houver um `-` em todas as posições do cartão.


```python
#@title Solução

from random import randint, sample

cartao = gerar_cartao()

imprimir_cartao(cartao)

n_tracos = 0
jogada = 0

numeros = sample(range(1, 76), 75) # seleciona a ordem aleatória dos números a sair para não haver reposição


while n_tracos != 25:
  n_random = numeros[jogada]
  print(f'O número é {n_random}.')

  for coluna in cartao:
    for i in range(0, len(cartao[coluna])):
      if cartao[coluna][i] == n_random:
        cartao[coluna][i] = '-'
        imprimir_cartao(cartao)
        n_tracos += 1

  jogada += 1

print(f'BINGO! Foram precisas {jogada} jogadas.')
```
