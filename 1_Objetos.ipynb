## Atribuição

A instrução mais básica em Python consiste em criar um objeto e dar-lhe um nome, ou seja, fazer com que um nome (variável) aponte para um determinado objeto (valor).

A esta instrução dá-se o nome de _**atribuição**_:

````python
x = 123     # cria o objeto 123 e dá-lhe o nome x

print(x)    # imprime o objeto referenciado pelo nome x
````

## Tipos de objetos

Podemos saber qual o *__tipo__* de objeto referenciado por um certo nome da seguinte forma:

````python
x = 123

type(x)
````

## Endereço de um objeto

Também podemos obter o _**endereço**_ de um objeto.

Este valor corresponde à posição onde determinado objeto é guardado na memória do computador.

````python
x = 123

id(x)
````

## _Aliasing_
É possível dar mais do que um nome ao mesmo objeto. Trata-se de uma operação chamada _**aliasing**_.

<details>
  <summary><font size="4">Explicação</font></summary>

Vemos que o endereço de `x` e `y` é o mesmo. Isto porque não foram criados dois objetos `45` diferentes, mas apenas um guardado num único local da memória. Este objeto pode é ser invocado por dois nomes distintos: `x` e `y`.

</details>

````python
x = 45
y = 45

print('id(x) = ', id(x))
print('id(y) = ', id(y))
````

## Baptismo duplo

Também podemos fazer este _**batismo duplo**_ na mesma linha:

````python
x = y = 45

print('id(x) = ', id(x))
print('id(y) = ', id(y))
````

## Mudança de objeto

No entanto, o mesmo nome não pode corresponder a mais do que um objeto ao mesmo tempo:

<details>
  <summary><font size="4">Explicação</font></summary>

Prevalece sempre a última instrução de atribuição. Quando se reutilizou o nome `x` para apontar para o novo objeto `43`, o objeto `20` fica sem nome e passa a estar inacessível, sendo destruído (apagado da memória).

</details>

````python
x = 20
x = 43

print(x)
````

## Destruir um objeto
Podemos também apagar explicitamente um objeto da memória:

<details>
  <summary><font size="4">Explicação</font></summary>

A partir do momento em em que apagamos o objeto da memória, já não conseguimos aceder a este objeto e obtemos um erro do tipo `NameError`.

</details>

````python
x = 43

del(x)
print(x)
````

# Input e Prints

Num programa é habitual que haja **interação com o utilizador**, para pedir dados ou colocar questões (variáveis iniciais). No final do programa também devemos ser capazes de devolver os resultados do programa. Isto é feito através de **intruções de entrada e saída**.



## Função `input()`

Para que um utilizador possa passar informação para o programa, existe o método `input()`.

````python
# Entrada de informação / input:
n = input('Introduza um valor inteiro: ') # pede um valor ao utilizador e guarda-o na variável n

# Saída de informação / output:
print('O valor fornecido foi', n)
````

## `input()` de valores numéricos

 O que é que vai acontecer?

 <details>
  <summary><font size="4">Explicação</font></summary>

Se fizermos esta operação verificamos que há um problema!

**Qualquer entrada a partir do método `input()` é interpretada como um objeto do tipo `string`.**

A solução é converter a `string` num inteiro, `int`, através do método `int()` (apenas é possível nos casos em que a `string` é composta por caracteres numéricos).

</details>

````python
n1 = int(input('Introduza um valor: '))  # pede um valor n1 ao utilizador
n2 = int(input('Introduza outro valor: '))  # pede um valor n2 ao utilizador

resultado = n1 + n2 # soma os dois valores
print('O resultado da soma é', resultado) # imprime o resultado
````

## Função `print()`

Como já vimos, o método `print()` é usado para imprimir informação no ecrã, como instruções para o utilizador, ou para devolver os resultados de um programa.

É também muito útil para fazer _debugging_ e encontrar erros no programa, fazendo `print()` de resultados / valores intermédios das variáveis.

Por defeito, o método `print()` faz sempre uma mudança de linha no fim:

````python
print('1')
print('2')
````

## Controlar o fim do `print()`

Há casos em que pode dar jeito eliminar esta mudança de linha e definir como pretendemos que acabe a impressão. Isto pode ser feito com a instrução `end`:

 <details>
  <summary><font size="4">Explicação</font></summary>

Neste caso, a é adicionado um espaço depois da primeira instrução de `print()`, em vez da habitual mudança de linha.

</details>

````python
print('1', end = ' ')
print('2')
````

## Formatação com `f-string`

A formatação com `f-string` é uma forma alternativa de imprimir resultados no ecrã e que, em alguns casos pode ser de mais fácil leitura do que usar a sintaxe clássica do `print()`.

Este tipo de formatação também permite mais controlo sobre o aspeto de como o _output_ é impresso no ecrã.



### Imprimir uma fatura

Da maneira que aprendemos, se quisermos imprimir os dados para uma fatura temos sempre um espaço entre o preço e o sinal `€`, que não conseguimos tirar.

Se imprimirmos com uma `f-string` conseguimos fazer esta correção de estética facilmente:

<details>
  <summary><font size="4">Solução</font></summary>

```
preco_u = 49.99
quantidade = 30

print(f"Preço unitário: {preco_u}€")
print(f'Quantidade: {quantidade} unidades')
print(f'Total: {quantidade * preco_u:.2f}€')
```
</details>

<details>
  <summary><font size="4">Explicação</font></summary>

Agora já podemos controlar melhor a formatação do que é devolvido ao utilizador!

A parte entre aspas é a **parte literal**, que é representada exatamente como a escrevemos. Onde queremos imprimir o valor de uma variável/expressão, devemos colocá-la dentro de `{}`.

Podemos ainda formatar a expressão com `:` antes de `}`. Por exemplo fazendo `:.2f` podemos formatar o número (`float`, `f`) para ser sempre impresso com 2 casas decimais.

</details>

````python
preco_u = 49.99
quantidade = 30

print(f'Preço unitário: {preco_u}€')
print('Quantidade:', quantidade, 'unidades')
print(f'Total: {quantidade * preco_u:.2f}€')
````

## 3. Energia Cinética

Escreva um programa que peça a massa, $m$, e a velocidade, $v$, de um objeto e calcule a energia cinética, com unidades.

$$E=\frac{1}{2}mv^2$$


````python
#@title Solução

m = float(input('Introduza a massa em kg: '))
v = float(input('Introduza a velocidade em m/s: '))

print(f'A energia cinética do objeto é {0.5 * m * v **2 :.2f} J.')
````

## 4. Fórmula Resolvente

Escreva um programa que determine as raízes para um polinómio de segundo grau do tipo $ax^2+bx+c=0$ através da fórmula resolvente:

$$x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}$$

````python
#@title Solução

a = float(input('Introduza a: '))
b = float(input('Introduza b: '))
c = float(input('Introduza c: '))

x_pos = (-b+(b**2-4*a*c)**0.5)/(2*a)
x_neg = (-b-(b**2-4*a*c)**0.5)/(2*a)

print(f'As soluções da equação {a}x^2 + {b}x + {c} = 0 são:')
print(f'  x = {x_neg}')
print(f'  x = {x_pos}')
````
