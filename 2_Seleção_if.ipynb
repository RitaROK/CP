# Lógica Booleana

Um programa faz decisões com base em **condições**. Uma condição que só pode ser **verdadeira** ou **falsa**.

Para isso, é necessário saber o **valor lógico** de objetos e expressões. Estas podem apenas assumir dois valores possíveis:

- **Verdadeiro:** `True` / `1`
- **Falso:** `False` / `0`

## Objetos booleanos

O valor lógico de um objeto pode ser obtido com a função `bool()`.

### `True`/`False` - Números

Um número é falso (`False`) se for igual a `0`. Caso contrário é verdadeiro (`True`).

````python
print(bool(6))
````

### `True`/`False` - Objetos

 Um objeto é `False` se estiver **vazio** ou se for `None`. Caso contrário, é `True`.

````python
print(bool('None'))
````

### Simplesmente `True`/`False` - Flags

Podemos simplesmente atribuir o valor de verdadeiro ou falso a uma variável.
Isto pode ser importante em alguns casos:

- Saber se um input é válido `valido = True` ou não `valido = Falso`;
- Saber se um jogo acabou `game_over = True` ou não `game_over = False`;
- Saber se há chocolate `chocolate = True` ou não `chocolate = False`;

Nestes casos é importante que o **nome da variável seja explicativo do seu próprio significado**.

**O estado booleano destas variáveis é suposto ser atualizado ao longo do programa** consoante o algoritmo.



````python
chocolate = True

print(type(chocolate))
````

# Seleção `if`...`elif`...`else`

A seleção `if` permite-nos escrever um programa que executa instruções diferentes para diferentes casos que dependem de uma certa condição ser `True`ou `False` (por exemplo, dependendo de um `input` introduzido pelo utilizador).

A sintaxe de um bloco de seleção `if`...`elif`...`else` tem a seguinte forma:

```
if <condição_1>:     # verifica se a condição_1 é verdadeira.
    <instruções_1>     # se sim, executa as instruções_1.
elif <condição_2>:   # (OPCIONAL!) senão, verifica a condição_2.
    <instruções_2>     # se a condição_2 for verdadeira, executa as instruções_2.
else:                # (OPCIONAL!) se nenhuma das condições anteriores for verdadeira:  
    <instruções_n>     # executa as instruções_n.
```

**A identação** (o espaço em branco (`  `)  no início das linhas de código) **é essencial!**

Esta é a forma como o Python percebe que instruções estão dentro de cada bloco de seleção! Este espaço pode corresponder a um espaço de parágrafo (`↹`), ou a um determinado número (constante) de espaços individuais.

<div>

</div>

## A clásula `if`

Para executar algo **se** uma condição for verdadeira.

Verificar **se** o aluno passou à cadeira e imprimir a nota.

````python
nota = 16

if nota >= 9.5:
  print('Parabéns, passaste a CP com', nota, 'valores.')

print('Acabou o programa.')
````

## A cláusula `else`

Para executar algo nas restantes ocasiões.

Imprimir também uma mensagem **caso** a o aluno não tenha passado à cadeira.

````python
nota = 9.4

if nota >= 9.5:
  print('Parabéns, passaste a CP com', nota, 'valores.')

else:
  print('Ainda não foi desta, mas o curso faz-se no recurso!')

print('Acabou o programa.')
````

## A cláusula `elif`

Quando, para além de verdadeiro / falso, queremos distinguir os resultados em mais do que duas categorias:

- 17 a 20: Defesa (ainda não sabemos a nota final).
- 9.5 a 17: Aprovado
- 0 a 9.5: Reprovado

**`if` versus `elif`:**
- Cada condição `if` é avaliada independentemente das restantes. **Gera múltiplas respostas.**
- Uma condição `elif` apenas é avaliada se as condições anteriores do mesmo bloco não se verificarem. **Gera uma única resposta.**



<details>
  <summary><font size="4">Solução</font></summary>

```
nota = 10

if nota >= 17:
  print('Parabéns: passaste a CP, mas ainda tens de ir a defesa para ter uma nota final.')

elif nota >= 9.5:
  print('Parabéns: passaste a CP com', nota)

else:
  print('Ainda não foi desta, mas o curso faz-se no recurso!')

</details>

````python
nota = 18

if nota >= 17:
  print('Parabéns, passaste a CP, mas ainda tens que ir subir o 17.')

elif nota >= 9.5:
    print('Parabéns, passaste a CP com', nota, 'valores.')

else:
    print('Ainda não foi desta, mas o curso faz-se no recurso!')

print('Acabou o programa.')
````

## Ninhos de `if`

Quando queremos especificar casos dentro de casos.

Informar **se** a nota fornecida é válida antes de a categorizar no tipo de nota.

if nota >= 0 and nota <= 20:

  else:

## 5. Pressão arterial

A pressão arterial é caracterizada por dois valores:

- **Pressão arterial máxima ou sistólica:** valor durante a contração do coração (sístole);
- **Pressão arterial mínima ou diastólica:** valor durante o relaxamento do coração (diástole);

Para efeitos de diagnóstico e tratamento consideram-se as seguintes categorias:

- **Normal:** pressão sistólica ≤ 120 e pressão diastólica < 80;
- **Pré-hipertensão:** 120 < pressão sistólica < 130 e pressão diastólica > 80;
- **Hipertensão de nível 1:** 130 ≤ pressão sistólica < 140 ou 80 ≤ pressão diastólica < 90;
- **Hipertensão de nível 2:** 140 ≤ pressão sistólica < 180 ou pressão 90≤ pressão diastólica < 120;
- **Crise hipertensiva:** 180 ≤ pressão sistólica ou 120 ≤ pressão diastólica.

Escreva um programa que peça a pressão sistólica e diastólica de um paciente e imprima a categoria do paciente.

Em caso de dúvida, considere sempre o caso mais crítico.

````python
#@title Solução

p_sistolica = float(input('Introduza a pressão arterial máxima (sistólica): '))
p_diastolica = float(input('Introduza a pressão arterial mínima (diastólica): '))

if p_sistolica >= 180 or p_diastolica >= 120:
  print('Paciente em crise hipertensiva.')

elif p_sistolica >= 140 or (p_diastolica >=90 and p_diastolica < 120):
  print('Paciente com hipertensão de nível 2.')

elif (p_sistolica >= 130 and p_sistolica <140) or (p_diastolica >= 80 and p_diastolica < 90):
  print('Paciente com hipertensão de nível 1.')

elif (p_sistolica > 120 and p_sistolica < 130) and (p_diastolica > 80):
  print('Paciente com pré-hipertensão.')

elif p_sistolica <= 120 and p_diastolica < 80:
  print('Paciente em situação normal.')
````

## 6. Anos bissextos

Os anos bissextos, nos quais o Fevereiro tem 29 dias, são:
  - anos divisíveis por 4, excepto quando são também divisíveis por 100 (1900 não é bissexto);

  - anos divisíveis por 400 (2000 é bissexto);

Escreva um programa que peça um ano ao utilizador e indique **se** ele é bissexto ou não.

````python
#@title Solução

ano = int(input('Introduza um ano: '))

# Vamos começar por assumir que o ano não é bissexto:
bissexto = False # criar uma "flag" que indica se o ano é bissexto (True) ou não (False).

if ano % 4 == 0 and ano % 100 != 0:  # se for divisível por 4, mas não por 100 é bissexto
  bissexto = True

if ano % 400 == 0:
  bissexto = True

if bissexto == True:
  print(f'O ano {ano} é bissexto.')

else:
  print(f'O ano {ano} não é bissexto.')
````

## 7. Índice de Massa Corporal
_(Ficha 4, Problema 3)_

Escreva um programa que defina uma função `imc(peso, altura)` que calcula o índice de massa corporal, **imc**, em unidades SI:

$$\text{imc}=\frac{\text{peso}}{\text{altura}^2}$$

O programa deve invocar a função com valores introduzidos pelo utilizador.

Apresente o resultado bem como a sua categoria, sabendo que o peso é:
- **Inferior ao normal** se `imc` < 18.5
- **Normal** se 18.5 ≤ `imc` < 25.0
- **Superior ao normal** se `imc` ≥ 25.

````python
#@title Solução

def imc(peso, altura):
    imc = peso / altura**2          # calcula o imc e depois decide o tipo
    if imc < 18.5:
        tipo = 'inferior ao normal.'
    elif imc < 25:
        tipo = 'normal.'
    else:
        tipo = 'superior ao normal.'
    return imc, tipo                # devolve o valor do imc e o tipo

peso = float(input('Qual o seu peso em kg? '))
altura = float(input('Qual a sua altura em m? '))

imc, tipo = imc(peso, altura)

print('O seu imc é', round(imc,2) , ', um valor', tipo)
````
