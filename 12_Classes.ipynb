# Classes

No `Python`, é possível criar uma classe de objetos novos, com métodos criados por nós. Vamos aprender como podemos fazê-los através da criação de uma classe de números racionais `Rational`.

## Definição

Uma classe é definida da seguinte forma:

```
class Rational: # definir a classe
    <métodos da classe>
```

Os métodos que queremos introduzir na classe são colocados no bloco identado.

## Método `__init__`

O método `__init__` é especial porque é aquele que vai definir cada objeto a partir dos seus atributos, isto é, as informações que definem cada objeto como único.

No caso de um número racional, os atributos são um numerador e um denominador.

`Self` representa o objeto concreto desta classe dentro do qual o método está a ser chamado.

````python
class Rational: # definir a classe
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d

print('Rational: ', type(Rational))
````

## Criar um objeto/instância da classe

A instrução class apenas define a classe, não cria nenhum objeto. Para criar um objeto do tipo Rational, temos que invocar a classe, como fazemos com as funções.

A partir deste momento, podemos aceder aos atributos do objeto fazendo `instância.atributo`

````python
class Rational: # definir a classe
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d

r1 = Rational(1,2) # criar um objeto da classe Rational (1/2)

print('Tipo r1: ', type(r1))

print('Numerador:', r1.numerator)

print('Denominador:', r1.denominator)
````

## Método `__repr__`

O método `__repr__` permite definir a forma com o `Python` representa o objeto quando fazemos `print()` de um objeto da classe:

````python
class Rational:
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d

    def __repr__(self):   # para definir como o Python vai imprimir os objetos da classe:
        return f'{self.numerator}/{self.denominator}'

r1 = Rational(1,2)

print('r1 =', r1)  # quando fazemos print, o python acede à string devolvida pelo método __repr__
````

## Outros Métodos

A partir de agora, podemos definir vários métodos que pretedemos que a classe tenha.


### Métodos de objeto único:

Para definir um método que calcule o valor (em `float`) de um determinado `Rational`, fazemos:

````python
class Rational:
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d
    def __repr__(self):   # para definir como o Python vai imprimir os objetos da classe:
        return f'{self.numerator}/{self.denominator}'

    def value(self):    # nome do método para obter o valor
        return float(self.numerator)/float(self.denominator) # fazer a divisão entre para obter o valor da fração


r1 = Rational(1,2)

print(r1, '=', r1.value()) # imprimir o resultado do método value
````

### Métodos de objetos múltiplios

Se quisermos multiplicar dois objetos `Rational`, temos que, para além do `self` (próprio objeto), dar-lhe um outro (`other)`:

````python
class Rational:
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d
    def __repr__(self):   # para definir como o Python vai imprimir os objetos da classe:
        return f'{self.numerator}/{self.denominator}'
    def value(self):    # nome do método para obter o valor
        return float(self.numerator)/float(self.denominator) # fazer a divisão entre para obter o valor da fração

    def multiply(self, other):  # nome do método para definir uma multiplicação: vamos precisar de dois objetos desta classe
        x = self.numerator * other.numerator
        y = self.denominator * other.denominator
        return Rational(x,y) # neste caso, o resultado da multiplicação é um Rational

r1 = Rational(1,2)
r2 = Rational(1,3)

print(r1,'x', r2, '=', r1.multiply(r2))
````

## Sobrecarga de Operadores

Há um conjunto de nomes especiais que nos permite aceder a um determinado método de uma classe a partir de um caracter especial. A isto dá-se o nome de **sobrecarga de operadores**. Por exemplo, podemos aceder ao método `multiply` com um `*` da seguinte forma:

````python
class Rational:
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d
    def __repr__(self):   # para definir como o Python vai imprimir os objetos da classe:
        return f'{self.numerator}/{self.denominator}'
    def value(self):    # nome do método para obter o valor
        return float(self.numerator)/float(self.denominator) # fazer a divisão entre para obter o valor da fração

    def __mul__(self, other):  # nome do método para definir uma multiplicação: vamos precisar de dois objetos desta classe
        x = self.numerator * other.numerator
        y = self.denominator * other.denominator
        return Rational(x,y) # neste caso, o resultado da multiplicação é um Rational

r1 = Rational(1,2)
r2 = Rational(1,3)

print(r1,'x', r2, '=', r1*r2)

print(2*3)
````

````python
class Rational:
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d
    def __repr__(self):   # para definir como o Python vai imprimir os objetos da classe:
        return f'{self.numerator}/{self.denominator}'
    def value(self):    # nome do método para obter o valor
        return float(self.numerator)/float(self.denominator) # fazer a divisão entre para obter o valor da fração
    def __mul__(self, other):  # nome do método para definir uma multiplicação: vamos precisar de dois objetos desta classe
        x = self.numerator * other.numerator
        y = self.denominator * other.denominator
        return Rational(x,y) # neste caso, o resultado da multiplicação é um Rational

    def __add__(self, other):  # nome do método para definir uma adição
        d = self.denominator * other.denominator # novo denominador
        n1 = self.numerator * other.denominator
        n2 = other.numerator * self.denominator
        return Rational(n1 + n2, d)

r1 = Rational(1,2)
r2 = Rational(1,3)

print(r1,'+', r2, '=', r1+r2)  # aplicar o método __add__
````

## Definir a comparação de objetos

Sem definição adicional, o `Python` só sabe que dois objetos da mesma classe são iguais se estes forem criados em conjunto.

Por isso, devemos definir na classe como devemos comparar estes objetos:

````python
class Rational:
    def __init__(self, n, d):   # definir o objeto com os seus atributos
        self.numerator = n
        self.denominator = d
    def __repr__(self):   # para definir como o Python vai imprimir os objetos da classe:
        return f'{self.numerator}/{self.denominator}'
    def value(self):    # nome do método para obter o valor
        return float(self.numerator)/float(self.denominator) # fazer a divisão entre para obter o valor da fração
    def __mul__(self, other):  # nome do método para definir uma multiplicação: vamos precisar de dois objetos desta classe
        x = self.numerator * other.numerator
        y = self.denominator * other.denominator
        return Rational(x,y) # neste caso, o resultado da multiplicação é um Rational
    def __add__(self, other):  # nome do método para definir uma adição
        d = self.denominator * other.denominator # novo denominador
        n1 = self.numerator * other.denominator
        n2 = other.numerator * self.denominator
        return Rational(n1 + n2, d)

    def __eq__(self, other):    # definir a igualdade de objetos ==
        if self.numerator == other.numerator and self.denominator == other.denominator:
        #if self.value() == other.value():
            return True
        else:
            return False

r1 = Rational(1, 3)
r2 = Rational(2, 6)

print(r1 == r2)
````

## Herança de classes

Podem definir-se subclasses a partir de uma classe existente (classe-mãe). Por default, a subclasse herda atributos e métodos da classe-mãe, pode pode ter outros adicionais:

Suponhamos que criamos uma classe `Person`, que define uma pessoa:

````python
class Person:

    def __init__(self, fname, lname): # definir uma classe Pessoa, definida por um primeiro e último nome
        self.firstname = fname
        self.lastname = lname

    def __repr__(self):
        return f'{self.firstname} {self.lastname}'

pessoa = Person('Albert', 'Einstein')

print(pessoa)
````

## Subclasse

Agora vamos criar uma subclasse Student desta classe-mãe que herda todos os métodos da classe mãe, a não ser que estejam definidos.

````python
class Person:

    def __init__(self, fname, lname): # definir uma classe Pessoa, definida por um primeiro e último nome
        self.firstname = fname
        self.lastname = lname

    def __repr__(self):
        return f'{self.firstname} {self.lastname}'

class Student(Person): # indica que é uma subclasse da classe Person

    def __init__(self, fname, lname, n): # estamos a mudar como definimos o objeto
        self.firstname = fname
        self.lastname = lname
        self.nestudante = n  # porque vamos acrescentar um número de estudante

    def email(self):    # um método para determinarmos o email do aluno
        return f'uc{self.nestudante}@student.uc.pt'

estudante = Student('Maria', 'Albertina', 2080811234)
print(estudante)
print(estudante.email())
````

## Reciprocidade

O que é importante aqui é que a instrução `print(estudante)` não dá erro. Isto porque, apesar de não estar definida na classe `Student`, ela herda este método da classe `Person`.

No entanto, a herança não funciona no sentido contrário: um objeto da classe `Person` não tem acesso aos métodos das suas subclasses:

````python
class Person:

    def __init__(self, fname, lname): # definir uma classe Pessoa, definida por um primeiro e último nome
        self.firstname = fname
        self.lastname = lname

    def __repr__(self):
        return f'{self.firstname} {self.lastname}'

class Student(Person): # indica que é uma subclasse da classe Person

    def __init__(self, fname, lname, n): # estamos a mudar como definimos o objeto
        self.firstname = fname
        self.lastname = lname
        self.nestudante = n  # porque vamos acrescentar um número de estudante

    def email(self):    # um método para determinarmos o email do aluno
        return f'uc{self.nestudante}@student.uc.pt'


pessoa = Person('Albert', 'Einstein')

print(pessoa.email())
````

## 1. Vetores

_(Ficha 9, Problema 1)_

Implemente uma classe de vetores `Vec` num espaço tridimensional.

### a) Operações

A classe deverá possuir métodos para o cálculo da norma, produto escalar, produto vetorial e ângulo entre os vetores.

Sobrecarregue os operadores `+` e `-` com o significado de adição e subtração vetorial e os operadores `|` e `&` com o significado de produto escalar e vetorial.

````python
#@title Solução

from math import acos, degrees

class Vec:
    def __init__(self, x, y, z):    # Inicialização
        self.x = x
        self.y = y
        self.z = z

    def __repr__(self):     # Representação
        return f'({self.x},{self.y},{self.z})'

    def norma(self):    # Norma
        return (self.x**2 + self.y**2 + self.z**2)**0.5

    def __or__(self, other):   # produto escalar a aceder por |
        return (other.x * self.x + other.y * self.y + other.z * self.z)

    def __and__(self, other):   # produto vetorial a aceder por &
        i = self.y * other.z - self.z * other.y
        j = - (self.x * other.z - self.z * other.x)
        k = self.x * other.y - self.y * other.x
        return Vec(i, j, k)

    def angulo(self, other):    # ângulo entre dois vetores
        return degrees(acos( (self|other) / float(self.norma() * other.norma()) ))

    def __add__(self, other):  # definir a soma de vetores
        i = self.x + other.x
        j = self.y + other.y
        k = self.z + other.z
        return Vec(i, j, k)

    def __sub__(self, other): # definir a subtração de vetores
        i = self.x - other.x
        j = self.y - other.y
        k = self.z - other.z
        return Vec(i, j, k)

vec_1 = Vec(1,0,0)

vec_2 = Vec(0,1,0)

print('Vetor de teste 1: ', vec_1)
print('Vetor de teste 2: ', vec_2)

print('Norma do vetor 1: ', vec_1.norma())

print('Produto escalar:', vec_1 | vec_2)

print('Produto vetorial:', vec_1 & vec_2)

print('Ângulo:', vec_1.angulo(vec_2), 'º')

print('Soma:', vec_1 + vec_2)

print('Subtração:', vec_1 - vec_2)
````

### b) Vetores unitários

A partir desta classe implemente uma subclasse de vetores unitários.

````python
#@title Solução

class Unit_vec(Vec):            # Herda da classe Vec

    def __init__(self, x, y, z):    # Para normalizarmos um vetor, temos que dividir cada coordenada pela norma
        vec_inicial = Vec(x, y, z)    # Inicializar o vetor inicial para depois podermos calcular a norma
        n = vec_inicial.norma()
        # definir as novas coordenadas do vetor normalizado:
        self.x = x / n
        self.y = y / n
        self.z = z / n

print('Vetor de teste: (2,0,0)')

print('Vetor unitário: ', Unit_vec(2,0,0))
````

## 2. Racionais

_(Ficha 9, Problema 2)_

Implemente uma classe de números racionais `Rat`. Um número racional é representado por uma fração com numerador e denominador inteiros.

### a) Operações

A classe deve implementar as operações elementares sobre os números racionais, fazendo a sobrecarga dos operadores habituais.

````python
#@title Solução

class Rat:
    '''This class implements rational numbers.'''

    def __init__(self, n, d):
        if d != 0 and type(n) == int and type(d) == int:      # podemos impedir que o objeto seja criado apenas se o denominador for maior que zero e tanto o numerador como o denominador sejam inteiros
            self.num = n
            self.den = d
        else:
            print('Objeto inválido.')

    def __repr__(self):
        if self.den == 1:
            return '{}'.format(self.num)
        else:
            return '{}/{}'.format(self.num, self.den)

    def value(self):
        return(self.num / self.den)

    def __add__(self, other):   # a + b
        return Rat(self.num * other.den + other.num * self.den, self.den * other.den)

    def __sub__(self, other):   # a - b
        return Rat(self.num * other.den - other.num * self.den, self.den * other.den)

    def __mul__(self, other):   # a * b
        return Rat(self.num * other.num, self.den * other.den)

    def __truediv__(self, other):   # a / b
        return Rat(self.num * other.den, self.den * other.num)

    def int_mul (self, b): # multiplicação por um número b
        return Rat(self.num * b, self.den)

    def __eq__ (self,other): # definir a igualdade de objetos
        n1 = self.num * other.den
        n2 = self.den * other.num
        if n1==n2:
            return True
        else:
            return False

r1 = Rat(1, 2)
r2 = Rat(1, 3)

print('Objetos de teste:', r1, r2)

print(r1, '=', r1.value())

print(r1, '+', r2, '=', r1 + r2)

print(r1, '-', r2, '=', r1 - r2)

print(r1, '*', r2, '=', r1 * r2)

print(r1, '/', r2, '=', r1 / r2)

print(r1, '==', r2, ':', r1 == r2)
````

### b) Forma Reduzida

A partir desta implemente uma subclasse de números racionais na forma reduzida.

````python
#@title Solução

from math import gcd

class Rat_red(Rat):     # vai ser uma subclasse de Rat
    def __init__(self, n, d):
        k = gcd(n, d) # encontrar o máximo divisor comum para reduzir a fração
        self.num = int(n / k)  # reduzir o numerador
        self.den = int(d / k)  # reduzir o denominador

num = int(input('Introduza o numerador: '))
den = int(input('Introduza o denominador: '))

r1 = Rat_red(num, den)  # gerar uma fração que não está reduzida para testar
print(f'O racional ({num},{den}) na forma reduzida é {r1}.')
````

## 3. Matrizes

_(Ficha 9, Problema 3)_

Implemente uma classe de matrizes 3 x 3 (`Mat`) que suporte algumas das operações habituais com matrizes, incluindo a soma, subtração, multiplicação e cálculo do determinante.

````python
#@title Solução

class Mat:
    def __init__(self, l1, l2, l3):     # a classe matriz vai receber as três linhas da matriz
        self.l1 = l1
        self.l2 = l2
        self.l3 = l3

    def __repr__(self):
        return f'|{self.l1[0]} {self.l1[1]} {self.l1[2]}| \n|{self.l2[0]} {self.l2[1]} {self.l2[2]}| \n|{self.l3[0]} {self.l3[1]} {self.l3[2]}|'

    def __add__(self, other):    # Soma de matrizes
        new_l1 = []
        new_l2 = []
        new_l3 = []
        for i in range(3): # Para cada coluna da matriz (são 3)
            new_l1.append(self.l1[i] + other.l1[i])   # obter o resultado da soma dos elementos correspondentes de cada matriz
            new_l2.append(self.l2[i] + other.l2[i])
            new_l3.append(self.l3[i] + other.l3[i])
        return Mat(new_l1, new_l2, new_l3) # transformar o resultado numa nova matriz

    def __sub__(self, other):
        new_l1 = []
        new_l2 = []
        new_l3 = []
        for i in range(3): # Para cada coluna da matriz (são 3)
            new_l1.append(self.l1[i] - other.l1[i])   # obter o resultado da soma dos elementos correspondentes de cada matriz
            new_l2.append(self.l2[i] - other.l2[i])
            new_l3.append(self.l3[i] - other.l3[i])
        return Mat(new_l1, new_l2, new_l3) # transformar o resultado numa nova matriz

    def __mul__(self, other):
        new_l1 = []
        new_l2 = []
        new_l3 = []
        for i in range(3):  # Para cada coluna, determinar o novo elemento de cada linha:
            new_l1.append(self.l1[0] * other.l1[i] + self.l1[1] * other.l2[i] + self.l1[2] * other.l3[i])
            new_l2.append(self.l2[0] * other.l1[i] + self.l2[1] * other.l2[i] + self.l2[2] * other.l3[i])
            new_l3.append(self.l3[0] * other.l1[i] + self.l3[1] * other.l2[i] + self.l3[2] * other.l3[i])
        return Mat(new_l1, new_l2, new_l3)  # transformar o resultado numa nova matriz

    def det(self):
        aei = self.l1[0] * self.l2[1] * self.l3[2]
        bfg = self.l1[1] * self.l2[2] * self.l3[0]
        cdh = self.l1[2] * self.l2[0] * self.l3[1]
        ceg = self.l1[2] * self.l2[1] * self.l3[0]
        bdi = self.l1[1] * self.l2[0] * self.l3[2]
        afh = self.l1[0] * self.l2[2] * self.l3[1]
        return  aei + bfg + cdh - ceg - bdi - afh

mat1 = Mat([1,2,3], [4,5,6], [7,8,9])
print(f'Matriz de teste: \n{mat1}')
print(f'Soma de matrizes: \n{mat1 + mat1}')
print(f'Subtração de matrizes: \n{mat1 - mat1}' )
print(f'Multiplicação de matrizes: \n{mat1 * mat1}')

print('Determinante:', mat1.det())
````

## 4. Tempo

Implemente uma classe de `Time` que suporte algumas das operações habituais com instantes temporais (horas, minutos e segundos), tais como a sua representação, a soma e a diferença entre instantes diferentes e a verificação se dois instantes são iguais.

````python
#@title Solução

class Time:
    def __init__(self, h, m, s, d=''): # o dia tem um valor default e é um atributo opcional (vai ser usado para as somas e dubtrações de horas)
        # só vamos deixar criar o objeto se a hora for válida
        if 0 <= h < 24 and 0 <= m < 60 and 0 <= s < 60:
            self.hora = h
            self.min = m
            self.seg = s
            self.dia = d

    def __repr__(self):
        return f'{self.hora}:{self.min}:{self.seg} {self.dia}'

    def __add__(self, other):

        # iniciar as variáveis
        new_min = 0
        new_h = 0
        dia = ''

        new_s = self.seg + other.seg # somar os segundos
        if new_s > 59: # se passarmos os 60 segundos, estamos no minuto a seguir
            new_s = new_s - 60
            new_min = 1

        new_min = self.min + other.min + new_min # somar os minutos, incluindo se vierem alguns dos segundos
        if new_min > 59: # se passarmos os 60 minutos, estamos na hora seguinte
            new_min = new_min - 60
            new_h = 1

        new_h = self.hora + other.hora + new_h # somar as horas, incluindo se vier alguma dos minutos
        dia=''
        if new_h > 23: # se passarmos as 23, estamos no dia seguinte
            new_h = new_h - 24
            dia = '(+1)' # para indicar que é o dia seguinte

        return Time(new_h, new_min, new_s, dia)

    def __sub__(self, other):

        # iniciar as variáveis
        new_min = 0
        new_h = 0
        dia = ''

        new_s = self.seg - other.seg # subtrair os segundos
        if new_s < 0: # se passarmos os 0 segundos, estamos no minuto anterior
            new_s = 60 + new_s
            new_min = -1

        new_min = self.min - other.min + new_min # subtrair os minutos, incluindo se vierem alguns dos segundos
        if new_min < 0: # se passarmos os 0 minutos, estamos na hora anterior
            new_min = 60 + new_min
            new_h = -1

        new_h = self.hora - other.hora + new_h # subtrair as horas, incluindo se vier alguma dos minutos
        dia=''
        if new_h < 0: # se passarmos as 0, estamos no dia anteior
            new_h = 24 + new_h
            dia = '(-1)' # para indicar que é o dia seguinte

        return Time(new_h, new_min, new_s, dia)

    def __eq__(self, other):
        if self.dia == other.dia and self.hora == other.hora and self.min == other.min and self.seg == other.seg:
            return True
        else:
            return False

time1 = Time(0,59,0)
time2 = Time(1,0,0)

print('Tempos para teste: ', time1, time2)
print('Soma:', time1 + time2)
print('Subtração:', time1 - time2)
print('Comparação:', time1 == time2)
````

## 5. Datas

_(Ficha 9, Problema 5)_

Escreva um programa que defina uma classe `Date` para tratamento de datas de calendário, com os atributos `day`, `month`, `year`. Considere o calendário Gregoriano para datas a partir do ano 1600. Utilize apenas operações numéricas sem recorrer a módulos específicos para tratamento de datas do `Python`.

Implemente os métodos seguintes:

- `valid(self)` que devolve `True` se a data for válida (ver nota para anos bissextos);
- `after(self)` que devolve a data do dia seguinte;
- `before(self)` que devolve a data do dia anterior;
- `__repr__(self)` para simplificar e uniformizar a apresentação de datas na formato `'2016/01/05'` ou `'####/##/##'` se a data for inválida;
- `diff(self, other)` que devolve o número de dias entres as duas datas;
- `weekday(self)` que devolva uma `string` com o dia da semana correspondente.

- **Nota:** os anos bisextos, nos quais o Fevereiro tem 29 dias, são anos divisíveis por 4, excepto anos divisíveis por 100, mas se o ano for divisível por 400, é outra vez bisexto. Assim, por exemplo, o ano 2000 é bisexto, mas 1900 ou 2100 não o são.

````python
#@title Solução

M30 = [4, 6, 9, 11]  # lista de meses com 30 dias
M31 = [1, 3, 5, 7, 8, 10, 12]  # lista de meses com 31 dias
Y_bis=[x for x in range(1600, 2500) if x % 4 == 0 and x % 100 != 0 or x % 400 == 0] # lista de anos bissextos

class Date:
    def __init__(self, d, m, y):
        self.day = d
        self.month = m
        self.year = y

    def valid(self): # a) devolve true se a data for válida

        if self.year > 1600 and 0 < self.month < 13: # confirmação do ano e do mês
            if self.month in M30 and 0 < self.day < 31:  # se for um mês de 30 dias e tiver entre 0 e 30 dias, é válido
                return True
            elif self.month in M31 and 0 < self.day < 32: # se for um mês de 31 dias e tiver entre 0 e 31 dias, é válido
                return True
            elif self.month == 2 and self.year in Y_bis and 0 < self.day < 30: # se for fevereiro, for um ano bissexto e tiver entre 0 e 29 dias, é válido
                return True
            elif self.month == 2 and self.year not in Y_bis and 0 < self.day < 29: # se for fevereiro, não for um ano bissexto, só pode ter entre 0 e 28 dias
                return True
            else:           # todos os outros casos são inválidos
                return False
        else:
            return False

    def after(self): # b) devolve a data do dia seguinte

        # temos que começar por determinar quantos dias tem o mês da nossa data
        if self.month in M30:
            dias_mes = 30
        elif self.month in M31:
            dias_mes = 31
        elif self.month == 2 and self.year in Y_bis:
            dias_mes = 29
        elif self.month == 2 and self.year not in Y_bis:
            dias_mes = 28

        new_dia = self.day + 1
        new_mes = self.month
        new_year = self.year

        if new_dia > dias_mes: # se passarmos o número de dias do mês:
            new_dia = 1 # passa a ser o dia 1
            new_mes = self.month + 1 # do mês seguinte
            if new_mes == 13: # se passar para o mês 13 (janeiro do ano a seguir)
                new_year = self.year + 1 # passamos para o ano seguinte
                new_mes = 1

        return Date(new_dia, new_mes, new_year)

    def before(self): # c) devolve a data do dia anterior
        new_dia = self.day - 1
        new_mes = self.month
        new_year = self.year

        if new_dia == 0: # se passarmos para o mês anterior temos que determinar qual o último dia desse mês
            new_mes = self.month - 1
            if new_mes in M30:
                new_dia = 30
            elif new_mes in M31:
                new_dia = 31
            elif new_mes == 2 and self.year in Y_bis:
                new_dia = 29
            elif new_mes == 2 and self.year not in Y_bis:
                new_dia = 28
            elif new_mes == 0: # se passarmos para antes de janeiro temos que passar para 31 de dezembro do ano anterior
                new_dia = 31
                new_mes = 12
                new_year = self.year - 1
        return Date(new_dia, new_mes, new_year)

    def __repr__(self): # d) representar a data
        if self.valid():
            return f'{self.year}/{self.month}/{self.day}'
        else:
            return '####/##/##'

    def diff (self, other): # e) devolve o nº de dias entre duas datas self - other
        n_dias = 0  # iniciar o contador de dias
        dia = self  # vamos andar para trás no tempo, dia a dia até as datas serem iguais, atualizando o contador
        while dia.day != other.day or dia.month != other.month or dia.year != other.year:
            dia = dia.before()
            n_dias += 1
        return n_dias

    def weekday(self): # f) devolve uma string com o dia da semana correspondente
        # o dia 1/1/1600 foi um sabado
        semana = ['sabado', 'domingo', 'segunda', 'terça', 'quarta', 'quinta', 'sexta']
        i = 0
        dia_ref = Date(1, 1, 1600)
        while dia_ref.day != self.day or dia_ref.month != self.month or dia_ref.year != self.year:
            dia_ref = dia_ref.after()
            i = i + 1
            if i == 7:
                i = 0
            dia_semana = semana[i]
        return dia_semana

date1 = Date(26,4,2025)
print(date1)
print(date1.valid())
print('Próximo dia:', date1.after())
print('Dia antes:', date1.before())
print('Diferença de dias: ', date1.diff(Date(1,1,2024)))
print(Date(26,4,2025).weekday())
````

### a) Classe

Defina uma classe `Ponto` que pretende definir pontos num sistema cartesiano a duas dimensões. Os atributos deste tipo de objeto devem ser as suas coordenadas $x$ e $y$. Esta classe deve ter definidos métodos que permitam:

- representação dos pontos na forma: $(x,y)$, ex: $(2,3)$;
- calcular a distância entre dois pontos, com a sobrecarga do operador `or` (`|`);
- um método `r_walk` que permite calcular um novo ponto depois de `n` deslocações de comprimento `d` em direções aleatórias (entre $0$ e $2\pi$), partindo do ponto inicial (`n` e `d` devem ser parâmetros de entrada neste método).

````python
#@title Solução

from random import random
from math import pi, sin, cos

class Ponto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self): # representação na forma (x,y)
        return f'({self.x}, {self.y})'

    def __or__ (self, other): # distância entre dois pontos
        return ( (self.x - other.x)**2 + (self.y - other.y)**2 )**0.5

    def r_walk (self, n, d): # calcular novo ponto depois de n deslocações de comprimento d numa direção aleatória
        new_x = self.x
        new_y = self.y

        for deslocacao in range(n):     # para cada iteração
            teta = random() * 2 * pi    # gerar o ângulo aleatório
            new_x = new_x + d * cos(teta) # somar à última posição o novo passo
            new_y = new_y + d * sin(teta) # somar à última posição o novo passo
        return Ponto(new_x, new_y) # devolver o ponto final com um objeto Ponto
````

- imprimir no ecrã o resultado do deslocamento aleatório de $n=20$ passos de grandeza $d=1$, para cada um dos pontos $p_1$ e $p_2$ e a distância final entre estes.

````python
#@title Solução

p1 = Ponto(2,5)
p2 = Ponto(-1,7)

print('A distância entre', p1, 'e', p2, 'é', p1 | p2)

print('O ponto final é de', p1, 'é',  p1.r_walk(20, 1))
print('O ponto final é de', p2, 'é',  p2.r_walk(20, 1))
````

## 7. Resistências Elétricas

_(Ficha 9, Problema 7)_

### a) Classe

Defina uma classe `Resistencia` que pretende definir resistências elétricas. O único atributo deste tipo de objeto deve ser o valor da resistência em ohm. Esta classe deve ter definidos os métodos:

- representação das resistências na forma: "valor" ohm (Ex: `53.6 ohm`);
- a associação em série de duas resistências, com sobrecarga do operador `add` (`+`)

$$R_{\text{série}}= R_1+R_2$$

- e a associação em paralelo, com sobrecarga do operador `or` (`|`)

$$\frac{1}{R_\text{paralelo}}=\frac{1}{R_1}+\frac{1}{R_2}$$

````python
#@title Solução

class Resistencia:

    def __init__(self, r):
        self.valor = r

    def __repr__(self):  # representação
        return f'{self.valor} ohm'

    def __add__(self, other):  # associação em série
        return Resistencia(self.valor + other.valor)

    def __or__ (self, other): # associação em paralelo
        return Resistencia( (self.valor * other.valor)/(self.valor + other.valor) )
````

### b) Criar objetos

Coloque instruções no programa que permitam:
- criar os objetos do tipo `Resistencia`, `R1` (`25 ohm`) e `R2` (`30 ohm`);
- imprimir no ecrã o resultado da associação em série desses dois objetos;
- imprimir no ecrã o resultado da associação em paralelo desses dois objetos.

````python
#@title Solução

R1 = Resistencia(25)
print('R1 =', R1)

R2 = Resistencia(30)
print('R2 =', R2)

print('Associação em série:', R1 + R2)
print('Associação em paralelo:', R1 | R2)
````

### c) Resistência Equivalente

Acrescente no programa instruções que permitam, a partir do ficheiro `resistencia.txt`, do tipo:

```
Resistencias
//: 50; 20; 89
//: 78; 21
+: 25; 45; 35
...
```

obter a associação equivalente de cada uma das linhas, onde `+` representa a associação em série e `//`, a associação em paralelo, usando o(s) método(s) adequados da classe anteriormente definida.

Os resultados destas operações devem ser guardados num ficheiro, designado por `resultado.txt` do tipo:

```
Resistência equivalente

//: 50; 20; 89 = 12.309820193637622 ohm
//: 78; 21 = 16.545454545454547 ohm
+: 25; 45; 35 = 105.0 ohm
...
```

````python
#@title Solução

from functools import reduce


f = open('resistencia.txt', mode = 'r')    # abrir o ficheiro de output no modo de leitura
linhas = f.readlines()  # lê todas as linhas do ficheiro, devolvendo-as numa lista de strings
f.close()

out = open('resultado.txt', mode = 'w')    # criar o ficheiro de output no modo de escrita
out.write('Resistência equivalente \n \n') # escrever o cabeçalho

for linha in linhas[1:]:                    # para cada linha
    linha = linha.strip('\n')               # retirar o parágrado
    output = linha + ' = '                  # copiar linha para depois escrever no ficheiro de output
    linha = linha.replace(';','')           # tirar o ;
    linha = linha.split(' ')                # separar as colunas pelos espaços

    lista_res = [Resistencia(float(x)) for x in linha[1:]] # gerar uma lista de objetos Resistencia

    if linha[0] == '//:':                               # reduzir a lista à função | entre resistências
        res = reduce(lambda r1, r2 : r1 | r2, lista_res)
    elif linha[0] == '+:':                              # reduzir a lista à função + entre resistências
        res = reduce(lambda r1, r2 : r1 + r2, lista_res)

    output = output + str(res) + '\n'   # acrescentar o resultado e o parágrafo
    out.write(output)                   # escrever a linha no ficheiro

out.close()
````
